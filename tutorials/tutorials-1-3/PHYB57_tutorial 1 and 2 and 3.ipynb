A string can be indexed, with the index starting from 0. This is also useful for slicing strings. Negative indices, starting at -1, index from the end of the string.

To slice a string, specify the starting index and the index AFTER the stopping index.


```python
print(a[4])      # output: o
print(a[-1])     # output: !
 
print(a[0:4])    # output: Hell
print(a[0:5])    # output: Hello
```

## LOGICAL EXPRESSIONS AND OPERATORS

A logical expression is an expression that has a truth value attached to it — it is either True or False. The boolean data type has only two values: 0/False and 1/True.


```python
print(1 < 2)    # output: True
print(1 > 2)    # output: False
```

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `print(1 < 2)    # output: True` | 判断 1 是否小于 2，结果是 True（真） |
| 2 | `print(1 > 2)    # output: False` | 判断 1 是否大于 2，结果是 False（假） |

We use comparison operators to construct logical expressions.

- `>` : Greater than
    >
- `>=` : Greater than or equal to
    >=
- `<` : Less than
    <
- `<=` : Less than or equal to
    <=
- `==` : Equal
    ==
- `!=` : Not equal
    !=

Note the difference between `=` and `==`. The first is used to assign values to variables, while the second is used to evaluate the equality of two values.


```python
b = 3
a = b         # assigns the value of b to a
print(a)      # output: 3

c = (a == b)  # assigns the boolean value of the logical expression to c
print(c)      # output: True
```

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `b = 3` | 把 3 存进 b |
| 2 | `a = b         # assigns the value of b to a` | 把 b 的值复制一份给 a |
| 3 | `print(a)      # output: 3` | 显示 a，就是 3 |
| 5 | `c = (a == b)  # assigns the boolean value of the logical expression to c` | ⚠️ 两个等号 == 是「判断是否相等」，把判断结果 True/False 存进 c |
| 6 | `print(c)      # output: True` | 显示 c，是 True |

In Python, a True expression has a value of 1, and a False expression has a value of 0.


```python
print((2 > 1) + (4 > 1) + (2 == 4))    # output: 2 (1 + 1 + 0)
```

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `print((2 > 1) + (4 > 1) + (2 == 4))    # output: 2 (1 + 1 + 0)` | Python 里 True 就是 1、False 就是 0，所以这是 1 + 1 + 0 = 2 |

Logical operators are used to create more complicated conditions involving two or more logical expressions.

- `and`: True if both expressions are true
    and
- `or`: True if at least one expression is true
    or
- `not`: Inverts the truth value of an expression
    not


```python
P = True
Q = False

and_operator = P and Q
or_operator = P or Q
not_operator = not P
combining_operators = P and not Q

print(and_operator, or_operator, not_operator, combining_operators)    # output: False True False True
```

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|

| 5 | `or_operator = P or Q` | or（或者）：有一个真就为真 → True |
| 6 | `not_operator = not P` | not（取反）：把 P 反过来 → False |
| 7 | `combining_operators = P and not Q` | 可以混着用：P and (not Q) → True |
| 9 | `print(and_operator, or_operator, not_operator, combining_operators)    # output: False True False True` | 把四个结果一起显示 |

## LISTS

Lists are defined using square brackets: `[item, item, item]`. The items in a list can be of any data type, and are ordered and changeable.

- ordered: `[1, 2, 3]` is NOT the same as `[3, 1, 2]`
    [1, 2, 3]
    [3, 1, 2]
- changeable: the items in the list can be modified

The position of an element in a list is determined by its index, starting at 0.


```python
int_list = [1, 3, 2, 7]
str_list = ["a", "b", "c", "a"]    # items can be repeated
mixed_list = [True, -4.1, "hi!", int_list, [5, 15]]   # different data types including other lists

print(int_list[2])        # output: 2; prints the 3rd item in int_list
print(mixed_list[4][0])   # output: 5; prints the 1st item in the 5th item [5, 15]
```

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `int_list = [1, 3, 2, 7]` | 用方括号建一个列表，里面装 4 个整数 |
| 2 | `str_list = ["a", "b", "c", "a"]    # items can be repeated` | 字符串列表。元素可以重复（两个 "a"） |
| 3 | `mixed_list = [True, -4.1, "hi!", int_list, [5, 15]]   # different data types including other lists` | 列表里可以混装不同类型，甚至可以把另一个列表装进去 |
| 5 | `print(int_list[2])        # output: 2; prints the 3rd item in int_list` | 取编号为 2 的元素，也就是第 3 个 |
| 6 | `print(mixed_list[4][0])   # output: 5; prints the 1st item in the 5th item [5, 15]` | 先取第 5 项（那个小列表 [5, 15]），再取它的第 1 项 → 5 |

A lot of operations that we used on strings also work on lists! There are also some specialised functions such as `append()` and `remove()` that are particular to lists.


```python
print(len(int_list))    # output: 4

concat_list = int_list + str_list
print(concat_list)      # output: [1, 3, 2, 7, "a", "b", "c", "a"]

str_list_slice = str_list[0:3]    # slices the list from index 0 to index 2 (stopping before index 3)
print(str_list_slice)             # output: ["a", "b", "c"]

str_list.remove("a")    # removes the first instance of "a" in the list
print(str_list)         # output: ["b", "c", "a"]

str_list.append("a")    # adds "a" to the end of the list
print(str_list)         # output: ["b", "c", "a", "a"]

str_list.pop(1)    # removes the element at index 1
print(str_list)    # output: ["b", "a", "a"]

str_list.insert(2, "c")    # inserts "c" at index 2
print(str_list)            # output: ["b", "a", "c", "a"]

str_list[3] = "k"    # sets the value of the item at index 3 to "k"
print(str_list)      # output: ["b", "a", "c", "k"]
```

**📖 这一格在做什么**

| 3 | `concat_list = int_list + str_list` | 两个列表首尾接起来，造出一个新列表 |
| 6 | `str_list_slice = str_list[0:3]    # slices the list from index 0 to index 2 (stopping before index 3)` | 切片：取编号 0 到 2 的元素（切到 3 之前停下） |
| 9 | `str_list.remove("a")    # removes the first instance of "a" in the list` | 按「值」删掉第一个等于 "a" 的元素 |
| 12 | `str_list.append("a")    # adds "a" to the end of the list` | 在列表末尾追加一个 "a" |
| 15 | `str_list.pop(1)    # removes the element at index 1` | 按「编号」删掉第 1 个元素（pop 还会把这个元素交出来） |
| 18 | `str_list.insert(2, "c")    # inserts "c" at index 2` | 在编号 2 的位置插入 "c"，后面的元素全部往后挪一格 |
| 21 | `str_list[3] = "k"    # sets the value of the item at index 3 to "k"` | 把编号 3 的位置直接改成 "k"（这是替换，不是插入） |

## TUPLES

Tuples are defined using round brackets: `(item, item, item)`. The items in a tuple can be of any data type, and are ordered and immutable.

- ordered: `(1, 2, 3)` is NOT the same as `(3, 1, 2)`
    (1, 2, 3)
    (3, 1, 2)
- immutable: the items of the tuple CANNOT be modified

The position of an element in a tuple is determined by its index, starting at 0.

Some list functions also work on tuples, including indexing, slicing, and `len()`. Tuples can also be concatenated to create new tuples. Tuples are especially useful when you want constant values in your code, or when you want to unpack results into multiple variables.


```python
my_tuple = (2, 4, 8, 16, 32)

print(len(my_tuple))    # output: 5
print(my_tuple[1])      # output: 4
print(my_tuple[1:3])    # output: (4, 8)

a, b, c, d, e = my_tuple    # automatically assigns each variable a value from the tuple in order
print(a, c, d)              # output: 2, 8, 16

new_item = (64,)          # the comma indicates that this is a tuple
my_longer_tuple = my_tuple + new_item
print(my_longer_tuple)    # output: (2, 4, 8, 16, 32, 64)
```

## SETS

Sets are defined using curly brackets: `{item, item, item}`. The items in a set can only be of immutable data types (so, not lists!); they are unordered and changeable.

- unordered: `{1, 2, 3}` is the same as `{3, 1, 2}`; this means that items cannot be repeated or indexed
    {1, 2, 3}
    {3, 1, 2}
- changeable: the items in the set can be modified

A set can be created from a list, a tuple, or a string using the `set()` function. The mathematical operations you might be used to performing on sets (union, intersection, etc.) can also be performed in Python.


```python
set_1 = set([1, 3, 5, 2, 5, 6, 2, 4, 2])
print(set_1)     # output: {1, 2, 3, 4, 5, 6}

set_2 = {2, 4, 6, 8, 12}
set_2.add(10)
set_2.remove(12)
print(set_2)     # output: {2, 4, 6, 8, 10}

union_set = set_1.union(set_2)    # returns set of items in either set without repeats
print(union_set)                  # output: {1, 2, 3, 4, 5, 6, 8, 10}

intersection_set = set_1.intersection(set_2)    # returns set of items common to both sets
print(intersection_set)                         # output: {2, 4, 6}

diff_set = set_1.difference(set_2)    # returns set of items in set_1 but not in set_2
print(diff_set)                       # output: {1, 3, 5}

symmdiff_set = set_1.symmetric_difference(set_2)    # returns set of items in either set but not both
print(symmdiff_set)                                 # output: {1, 3, 5, 8, 10}

print(set_1.issubset(union_set))         # output: True
print(set_1.issuperset(symmdiff_set))    # output: False
```

## DICTIONARIES

## IF/ELSE STATEMENTS

By default, Python executes code sequentially from top to bottom. Conditional statements allow our program to make decisions and execute specific blocks of code only when certain conditions are met.

In Python, code blocks under a condition are defined using a colon `:` and indentation (standard: 4 spaces).

- `if`: evaluates a condition and runs the indented block if the condition is `True`
    if
    True
- `elif` (short for "else if"): checked only if all preceding `if` and `elif` conditions were `False`
    elif
    if
    elif
    False
- `else`: runs if none of the preceding conditions were met

```python
temp = -5

if temp <= 0:
    state = "solid"
elif temp < 100:
    state = "liquid"
else:
    state = "gas"

print(f"At {temp}°C, water is a {state}.")    # output: At -5°C, water is a solid.

# CHECK: what would happen if we used three separate 'if' statements instead of 'elif' and 'else'?
```

## FOR LOOPS

A `for` loop is used to iterate over a sequence (such as a list, string, tuple, dictionary, or range of numbers).

To loop over a list/string/tuple/etc.:

- `for variable in data_structure`: in Python, the loop variable automatically takes on the value of each element in the sequence, one by one
    for variable in data_structure
- `for idx, variable in enumerate(seq)`: the loop variable returned is a tuple of the element's index and value
    for idx, variable in enumerate(seq)

These are a few ways to generate sequences of numbers:

- `range(stop_val)`: generates numbers from `0` up to (but not including) `stop_val`
    range(stop_val)
    0
    stop_val
- `range(start_val, stop_val, step_val)`: custom starting point `start_val`, stopping point `stop_val`, and step size `step_val` (default = 1)
    range(start_val, stop_val, step_val)
    start_val
    stop_val
    step_val

Some other useful loop commands:

- `break`: exits the loop immediately and executes the rest of the code
    break
- `continue`: skips the rest of the current iteration and jumps to the next one
    continue


```python
# looping over a list
numbers = [1, 2, 3, 4, 5]
for num in numbers:
    print(f"{num} squared: {num**2}")

# accumulating values over a range
total = 0
for i in range(1, 6):   # iterates over 1, 2, 3, 4, 5 (stops before 6)
    total += i          # shorthand for total = total + i

print("Sum from 1 to 5:", total)    # output: Sum from 1 to 5: 15
```

| 7 | `total = 0` | 累加器先清零 |

## WHILE LOOPS

A `while` loop repeatedly executes a block of code as long as a specified condition remains `True`.

Generally:

- use `for` loops when you know the sequence or the number of iterations in advance
    for
- use `while` loops when looping depends on an evolving condition (e.g. numerical convergence, user input, reaching a threshold, etc.)
    while

For while loops, always make sure the condition will eventually become `False` (or encounter a `break`), otherwise your code will be stuck in an infinite loop.


```python
# simulating radioactive decay: half-life of 5 years
mass = 80.0       # initial mass in grams
half_life = 5.0   # years per half-life
years = 0.0

while mass >= 10.0:
    mass = mass / 2.0
    years += half_life
    print(f"After {years} years: {mass:.1f} g remaining")

print(f"Decayed below 10 g in {years} years.")

# CHECK: what would happen if we commented out the line 'years += half_life'?
```

## FUNCTIONS

A function is a reusable block of code that takes inputs (called parameters or arguments), performs actions, and returns an output.

A function is defined using the `def` keyword: `def function_name(param1, param2):`

- `return` sends a value back to the caller so it can be saved in a variable or used in subsequent calculations
    return
- `return` vs. `print()`: `print()` only displays text on the screen; `return` gives the resulting data back to your program
    return
    print()
    print()
    return

A function can have default parameters, which have set values unless specifically modified when the function is called (e.g., `g=9.81`).

It is good practice to use docstrings when writing functions. Docstrings are multi-line comments written using triple quotes (`"""..."""`) immediately under the `def` line to explain what the function does, what inputs it expects, and what outputs it returns.


```python
import math

# function to calculate kinetic energy
def kinetic_energy(mass, vel):
    """
    Calculates kinetic energy: KE = 0.5 * m * v^2
    
    INPUTS:
    mass (float): mass in kg
    vel (float): velocity in m/s
    
    OUTPUTS:
    ke (float): kinetic energy in J
    """
    ke = 0.5 * mass * (vel ** 2)
    return ke

# function to calculate range of a projectile, using a default constant for g
def projectile_range(v0, angle_deg, g=9.81):
    """
    Calculates horizontal range of a projectile.
    
    INPUTS:
    v0 (float): initial velocity in m/s
    angle_deg (float): angle of projectile launch in degrees
    
    OUTPUTS:
    r (float): range of projectile in km, rounded to 2 decimal places
    """
    angle_rad = math.radians(2 * angle_deg)
    r = (v0 ** 2) * math.sin(angle_rad) / g
    return round(r, 2)

# calling kinetic_energy and storing the returned value
energy = kinetic_energy(2.0, 3.0)
print("Kinetic Energy:", energy, "J")    # output: Kinetic Energy: 9.0 J

# calling projectile_range with default g
range_earth = projectile_range(20.0, 45.0)

# calling projectile_range with Mars gravity (g=3.71)
range_mars = projectile_range(20.0, 45.0, g=3.71)

print(f"Range on Earth: {range_earth} m")    # output: Range on Earth: 40.77 m
print(f"Range on Mars: {range_mars} m")      # output: Range on Mars: 107.82 m
```

Functions can be defined inside other function; these are called nested functions.

Why use nested functions?

1. helper code that is only meaningful inside the outer function
2. the inner function has access to variables in the outer function's scope
3. an outer function can configure and return a specialized inner function


```python
# nested helper function
def solve_quadratic(a, b, c):
    """
    Finds the real roots of ax^2 + bx + c = 0 using a nested discriminant helper.
    
    INPUTS:
    a (float): leading coefficient of function
    b (float): middle coefficient of function:
    c (float): constant of function
    
    OUTPUTS:
    (root1, root2): tuple of real roots
    """
    def get_discriminant():
        # accesses a, b, and c directly from the enclosing scope
        return b**2 - 4*a*c
    
    d = get_discriminant()
    if d < 0:
        return "No real roots"
    root1 = (-b + d**0.5) / (2*a)
    root2 = (-b - d**0.5) / (2*a)
    return (root1, root2)

print(solve_quadratic(1, -5, 6))    # output: (3.0, 2.0)
print(solve_quadratic(1, 2, 5))     # output: No real roots

# specialised inner function
def make_power_function(power):
    """
    Creates a customized function that raises inputs to the specified power.
    
    INPUTS:
    power (float): specified power of custom function
    
    OUTPUTS:
    power_func (function): function that raises input to specified power
    """
    def power_func(x):
        return x ** power  # remembers 'power' even after make_power_function finishes
    return power_func

square = make_power_function(2)
cube = make_power_function(3)

print(square(5))    # output: 25
print(cube(3))      # output: 27

# CHECK: can you call 'get_discriminant()' directly from outside 'solve_quadratic()'?
```

A lambda function is a small, often unnamed function defined in a single line using the `lambda` keyword: `lambda arguments: expression`.

It can take any number of arguments, separated by commas, but it can contain **only one expression**, which is automatically evaluated and returned (no `return` keyword is used).

It is best used for short, throwaway operations, especially as key functions in `sorted()`, `min()`, `max()`, or with `filter()` and `map()`. If logic requires loops or multiple lines, always define a regular function using `def` instead.


```python
# regular function vs. lambda function
# def add(x, y):
#     return x + y
add = lambda x, y: x + y
print(add(4, 7))    # output: 11

# converting Celsius to Fahrenheit
c_to_f = lambda c: (c * 9/5) + 32
print(f"20°C is {c_to_f(20)}°F")    # output: 20°C is 68.0°F

# sorting complex data (list of tuples) using lambda keys
students = [("Alice", 88), ("Bob", 92), ("Charlie", 78), ("Diana", 95)]

by_name = sorted(students, key=lambda item: item[0])
print("Sorted by name:", by_name)

by_score = sorted(students, key=lambda item: item[1])
print("Sorted by score:", by_score)
```
