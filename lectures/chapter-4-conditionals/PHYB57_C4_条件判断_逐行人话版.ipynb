# §1 · if / else 与缩进（P1）

## 课件原文（P1）

- **Branching Statements Overview:**
  - An **If-Else Statement** (or simply If Statement) is used to **control the flow of code**.
  - It executes a block of code **only if** a given logical condition is met.
  - If the condition is true, a specific code block is executed; otherwise, the program **skips to the next block**.
- **Basic Syntax:**

```python
if condition:
    code_block
else:
    alternative_code_block
```

- **Indentation:**
  - All lines within the `if` or `else` blocks **must have the same level of indentation**, typically **four spaces**.

## 人话版

**if 语句就是一句话：「如果……就……否则……」**

```python
if condition:            # <- 问一个问题（条件是 True 还是 False？）
    code_block           # <- True 就执行这一块
else:
    alternative_code_block   # <- False 就执行这一块
```

**三个组成部分**

| 部分 | 关键点 |
|---|---|
| `if condition:` | ⚠️ **结尾那个冒号不能忘** |
| 缩进的代码块 | ⚠️ **缩进就是 Python 的「大括号」** —— 它决定了「哪些行属于 if」 |
| `else:` | **可选的**。不需要「否则」的时候可以不写 |

**缩进：Python 最独特的语法规则**

其他语言用大括号表示「这一块」：

```c
if (condition) {
    code_block;          // C/Java/JavaScript 的写法
}
```

**Python 用缩进**：

```python
if condition:
    code_block           # Python 的写法：缩进就是「块」
```

| 规则 | 说明 |
|---|---|
| 同一个块里缩进必须**完全一致** | 都是 4 个空格，或者都是 1 个 Tab，**不能混** |
| 约定用 **4 个空格** | PEP 8（官方风格指南）的规定 |
| 缩进结束 = 块结束 | 一旦不缩进了，就说明「这个 if 管到这里为止」 |

> ⚠️ **为什么缩进这么容易出错？**
> 因为**空格和 Tab 看起来一样，但 Python 认为它们不同**。
> 同一个块里混用，会报 `TabError` 或者 `IndentationError`。
>
> **解决办法**：在 Jupyter 里**永远用 Tab 键**（它会自动插入 4 个空格），
> 或者设置编辑器「按 Tab 插入空格」。**绝对不要手动敲空格再混着敲 Tab。**

## §1.1 课件那段语法骨架

课件给的是**骨架**（`condition`、`code_block` 都是占位符），不能直接跑。
下面把它填成能跑的真实代码。

````python
temperature = 25

if temperature > 30:
    print('好热')
else:
    print('还行')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `temperature = 25` | 准备一个数据 |
| 3 | `if temperature > 30:` | **问一个问题**：25 > 30 吗？答案是 `False` |
| 4 | `print('好热')` | ⚠️ **缩进了 4 格** → 属于 if 的块。因为条件是 False，**这一行被跳过** |
| 5 | `else:` | 「否则」 |
| 6 | `print('还行')` | 缩进了 4 格 → 属于 else 的块。**执行这一行** |

**输出只有一行**：`还行`。

> 💡 **注意 `if` 和 `else` 是「二选一」，不是「各来一次」。**
> **两个块里必然执行且只执行一个。** 这是初学者最容易搞错的地方 ——
> 有人以为 `if` 不成立时两个都跳过，其实一定会走 `else`。

## §1.2 🧪 把四种情况都跑一遍

光看一种情况不够。下面用一个循环把「条件为真」和「条件为假」都跑出来 ——
顺便让你体会 **if 和循环怎么配合**（C5 会正式讲循环）。

````python
def describe(temp):
    if temp > 30:
        return '好热'
    else:
        return '还行'


print('--- 同一段 if-else，喂不同的数据 ---')
for t in [35, 30, 25, -5]:
    result = describe(t)
    why = f'{t} > 30 是 {t > 30}'
    print(f'  temp = {t:>4}  ->  {result:<4}   （因为 {why}）')

print()
print('--- 关键：条件是 False 时，if 那一块完全不被执行 ---')
log = []


def spy(temp):
    log.clear()
    if temp > 30:
        log.append('走了 if 分支')
    else:
        log.append('走了 else 分支')
    return log


print('  spy(35) ->', spy(35))
print('  spy(25) ->', spy(25))
print('  spy(30) ->', spy(30), '  <- 30 > 30 是 False，所以走 else')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-5 | `def describe(temp):` | 把 if-else 包进函数（C3 学的），这样能反复调用 |
| 9-11 | `for t in [35, 30, 25, -5]:` | 喂四个数据。⚠️ **注意 `30` 这个边界值** |
| 10 | `f'{t} > 30 是 {t > 30}'` | 把**条件本身的计算结果**打出来，这样你能看见 Python 到底怎么判的 |
| 19-26 | `spy` 函数 | 用一个列表记录「到底走了哪个分支」，**把执行路径显式记录下来** |

**边界值的处理值得专门看一眼：**

| 输入 | `temp > 30` | 走哪个分支 |
|---|---|---|
| 35 | True | if |
| **30** | **False** | **else** |
| 25 | False | else |
| −5 | False | else |

> ⚠️ **`temp = 30` 走的是 `else`！** 因为 `30 > 30` 是 `False`。
> **`>` 不包含等于。** 想要包含 30，得写 `>=`。
>
> 💡 **这种「边界差一」的错误（off-by-one）是编程里最常见、最难查的 bug。**
> 写任何条件判断时，**都问自己一句：「边界值走哪条路？这是我要的吗？」**
>
> 🧪 建议：每写一个 `if`，都像上面这样**把边界值喂进去跑一遍**。

## §1.3 🧪 缩进实验：Python 怎么知道「块」在哪里

这是本节最该亲手做一次的实验。

````python
score = 75

print('=== 实验 A：else 里的两行 ===')
if score >= 60:
    print('  及格了')
else:
    print('  没及格')
    print('  这行也在 else 里面（同样缩进了 4 格）')

print()
print('=== 实验 B：把第二行「顶格」，它就不属于 else 了 ===')
if score >= 60:
    print('  及格了')
else:
    print('  没及格')
print('  这行没有缩进 -> 不属于 if/else，永远都会执行')

print()
print('=== 实验 C：嵌套缩进（if 里面还有 if）===')
if score >= 60:
    print('  及格了')
    if score >= 90:
        print('    而且是优秀（缩进 8 格）')
    else:
        print('    但不是优秀（缩进 8 格）')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-9 | 实验 A | `else:` 下面有**两行**，都缩进 4 格 → **两行都属于 else** |
| 13-17 | 实验 B | ⚠️ 最后那行 `print(...)` **顶格写** → 它**不属于** `if/else`，所以**无论条件真假都会执行** |
| 20-25 | 实验 C | **嵌套**：`if` 里面再写 `if`，内层缩进 **8 格** |

**实验 B 的输出来自 score = 75，走的是 if 分支，所以你会看到：**

```
=== 实验 B：把第二行「顶格」，它就不属于 else 了 ===
  及格了
  这行没有缩进 -> 不属于 if/else，永远都会执行
```

> 💡 **「这行没有缩进」这句话本身就是本节的核心。**
>
> **在 Python 里，缩进不是「排版好看」，它是语法。**
> 同样两行代码，缩进不同，**含义完全不同**：
>
> | 缩进 | 含义 |
> |---|---|
> | 4 格 | 属于 `if` 的块 |
> | 4 格（在 else 下） | 属于 `else` 的块 |
> | 0 格（顶格） | **不属于任何分支**，无条件执行 |
> | 8 格 | 属于内层嵌套的块 |
>
> ⚠️ **这条规则是所有「Python 代码跑出来和我想的不一样」问题里最常见的原因。**
> **遇到诡异的行为，先检查缩进。**

# §2 · elif：多个分支（P2）

## 课件原文（P2）

- **Extended If-Else:** You can add multiple conditions using `elif` (short for **else-if**).
- Python checks each condition **in sequence**, and once it finds a **True** condition, it **executes the corresponding block and exits**.
- **Syntax:**

```python
if condition_1:
    code_block_1
elif condition_2:
    code_block_2
elif condition_3:
    code_block_3
else:
    default_code_block
```

- **Key Points:**
  - The `elif` statement allows checking additional conditions.
  - You can have **as many `elif` statements as needed**, but only **one `else`** statement.
  - **Once a True condition is found, no further conditions are evaluated.**

## 人话版

**`elif` 就是 `else` + `if` 的缩写：**

```python
if 条件1:
    ...
elif 条件2:        # <- 「否则，如果……」
    ...
elif 条件3:
    ...
else:              # <- 「否则」（兜底）
    ...
```

**三条规则（课件都说了，逐条解释）**

| 规则 | 意思 | 后果 |
|---|---|---|
| `elif` 可以有很多个 | 想加多少分支就加多少 | 分段函数随便写 |
| `else` 只能有**一个**，而且是**可选的** | 兜底只能有一个 | 也可以一个都不写 |
| **命中即停** | 一旦某个条件为真，**后面的条件根本不看** | ⚠️ **顺序会影响结果！** |

> ⚠️ **「Once a True condition is found, no further conditions are evaluated」
> 是本节最重要的一句话。**
>
> 它的意思是：**`elif` 链的判定是「短路」的**（和 C1 §7.2 的 `and`/`or` 一个道理）。
> 这带来一个极其常见的 bug：**如果你把「宽的条件」写在前面，
> 「窄的条件」就永远轮不到。**
>
> 下面 §2.2 会实测这个坑。

**和 `if` + `if` 的区别（超级重要）**

```python
# 写法 A：elif 链 —— 只会执行【一个】分支
if score >= 90:
    grade = 'A'
elif score >= 60:
    grade = 'P'
else:
    grade = 'F'

# 写法 B：独立的 if —— 可能执行【多个】分支
if score >= 90:
    print('优秀')
if score >= 60:
    print('及格')
```

| 写法 | 会执行几个分支 |
|---|---|
| `if / elif / else` | **恰好一个**（互斥） |
| `if / if / if` | **0 个到全部**（互相独立） |

> 💡 **判断方法**：问自己「这些条件**可能同时成立**吗？」
> - 可能同时成立、且要都执行 → 用独立的 `if`
> - 互斥、只要挑一个 → 用 `elif`
>
> 成绩分级是**互斥**的（一个分数只对应一个等级）→ 用 `elif`。

## §2.1 把课件的骨架填成能跑的分级程序

````python
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'


print('--- 逐个分数试一遍 ---')
for s in [95, 90, 85, 72, 60, 59, 0]:
    print(f'  {s:>3} 分 -> {grade(s)}')

print()
print('--- 注意边界：每个分数只落进一个分支 ---')
for s in [89.9, 90.0, 90.1]:
    print(f'  {s:>5} -> {grade(s)}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2 | `if score >= 90:` | 第一个分支 |
| 4-8 | `elif score >= 80:` ... | 依次往下 |
| 9-10 | `else:` | **兜底**：前面全不成立时走这里 |
| 14-15 | 循环测试 | 7 个代表性分数 |
| 19-20 | 边界测试 | ⚠️ **盯着 90 附近看** |

**边界测试的结果，请特别注意：**

| 分数 | 走哪个分支 | 为什么 |
|---|---|---|
| 89.9 | `elif score >= 80` -> B | 89.9 >= 90 是 False |
| **90.0** | `if score >= 90` -> A | 正好卡在边界上，`>=` 包含等于 |
| 90.1 | A | |

> 💡 **注意每个分支的写法都是从大到小排的（90 -> 80 -> 70 -> 60）。**
> **这个顺序不是随便定的**，理由见下一格。
>
> ⚠️ **另一个细节：这个函数用了一串 `return`，没用 `grade = ...`。**
> `return` 会**立刻结束函数** —— 所以它天然就是「命中即停」，
> **连 `elif` 都可以换成 `if`**（因为 return 之后就出去了）。
> 这里写 `elif` 是为了表达「这些条件是互斥的」这层意思，**更清楚。**

## §2.2 ⚠️ 顺序的陷阱：宽的条件写在前面会怎样

这是 `elif` 链最容易踩的坑。**下面故意把顺序写反。**

````python
def grade_wrong(score):
    """故意把条件写反：从宽到窄"""
    if score >= 60:
        return 'D（及格）'
    elif score >= 70:
        return 'C'
    elif score >= 80:
        return 'B'
    elif score >= 90:
        return 'A'
    else:
        return 'F'


def grade_right(score):
    """正确：从窄到宽"""
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D（及格）'
    else:
        return 'F'


print('分数   错误版本        正确版本')
print('-' * 45)
for s in [95, 85, 75, 65, 55]:
    print(f'{s:>4}   {grade_wrong(s):<14} {grade_right(s)}')
print()
print('错误版本里，只要 >= 60 就立刻返回了，后面三个 elif 永远轮不到！')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-11 | `grade_wrong` | ⚠️ **第一个条件 `score >= 60` 是最宽的** —— 95 分也满足它 |
| 14-23 | `grade_right` | 从窄到宽：先问「是不是 >= 90」，再问「是不是 >= 80」…… |
| 28-30 | 对照表 | 同一批分数，两个函数给出**不同答案** |

**跑出来是这样：**

| 分数 | 错误版本 | 正确版本 |
|---|---|---|
| 95 | `D（及格）` | `A` |
| 85 | `D（及格）` | `B` |
| 75 | `D（及格）` | `C` |
| 65 | `D（及格）` | `D（及格）` |
| 55 | `F` | `F` |

> ⚠️ **注意：错误版本「不会报错」，它只是安静地给出错误答案。**
> 这是最难查的一类 bug —— **程序跑得好好的，就是结果不对。**
>
> 💡 **两条实用规则**：
>
> 1. **写 `elif` 链时，条件要么「互斥」，要么「从窄到宽」排列。**
>    上面这个例子里，错误版本的问题是它**压根不互斥**（95 分同时满足所有条件）。
> 2. **写完 `elif` 链，一定要拿边界值各试一遍**（就像上一格做的那样）。

## §2.3 🧪 `elif` vs 独立 `if`：一个直观对照

````python
score = 85

print('=== 写法 A：elif 链（互斥，只执行一个）===')
if score >= 90:
    print('  A 分支')
elif score >= 80:
    print('  B 分支')
elif score >= 60:
    print('  C 分支')

print()
print('=== 写法 B：独立 if（互相独立，可能都执行）===')
if score >= 90:
    print('  优秀')
if score >= 80:
    print('  良好')
if score >= 60:
    print('  及格')

print()
print('  -> 写法 A 只打印了一行；写法 B 打印了两行')
print('  -> 85 分「既良好，又及格」—— 两个说法都对，所以都打印')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 3-9 | `if ... elif ... elif` | 85 分：第一个条件 False，第二个 True → **打印 B 分支后立刻跳出整条链** |
| 12-17 | 三个独立的 `if` | 85 分：第一个 False，第二个 True（打印），第三个 **也 True（也打印）** |

**两种写法的输出对比：**

| 写法 | 输出行数 |
|---|---|
| `elif` 链 | **1 行**（B 分支） |
| 独立 `if` | **2 行**（良好、及格） |

> 💡 **没有哪个「更对」，只有哪个「更符合你的意图」。**
>
> | 你的意图 | 该用 |
> |---|---|
> | 这些情况**互相排斥**，只需确定是哪一个 | `if / elif / else` |
> | 这些情况**可以同时成立**，每一条都要处理 | 独立的 `if` |
> | 需要**给所有情况打标签**（比如「及格」和「良好」两个标签都要） | 独立的 `if` |
>
> ⚠️ **判断方法**：把条件都写出来，问「**同一个输入能不能同时满足两个条件？**」
> 能 → 独立 `if`；不能 → `elif`。

# §3 · 实例：恒温器 `my_thermo_stat`（P3）

## 课件原文（P3）

- **Task:** Write a function `my_thermo_stat(temp, desired_temp)` that adjusts a thermostat based on the temperature difference.
- **Functionality:**
  - If the current temperature is **more than 5 degrees lower** than the desired temperature, return **"Heat"**.
  - If it's **more than 5 degrees higher**, return **"AC"**.
  - Otherwise, return **"Off"**.

```python
def my_thermo_stat(temp, desired_temp):
    if temp < desired_temp - 5:
        return 'Heat'
    elif temp > desired_temp + 5:
        return 'AC'
    else:
        return 'Off'
```

**Sample Usage**

```python
print(my_thermo_stat(65, 75))   # Output: Heat
print(my_thermo_stat(75, 65))   # Output: AC
print(my_thermo_stat(65, 63))   # Output: Off
```

## 人话版

**这是本章唯一一个完整的、有实际意义的函数。** 它值得逐行拆。

**任务翻译**

| 课件英文 | 人话 | 代码 |
|---|---|---|
| more than 5 degrees **lower** | 比目标温度**低超过 5 度** | `temp < desired_temp - 5` |
| more than 5 degrees **higher** | 比目标温度**高超过 5 度** | `temp > desired_temp + 5` |
| Otherwise | 其余所有情况 | `else` |

**为什么不写成 `desired_temp - temp > 5`？**

两种写法**完全等价**，但课件选了 `temp < desired_temp - 5`。这个选择很讲究：

| 写法 | 读起来 |
|---|---|
| `temp < desired_temp - 5` | 「当前温度 **低于**（目标 − 5）」 |
| `desired_temp - temp > 5` | 「（目标 减 当前）**大于 5**」 |

**第一种把「谁是主语」放在最左边**，更接近自然语言 ——
"if the current temperature is more than 5 degrees lower than the desired temperature"。
**照着英文句子从左往右翻译，就得到课件的写法。**

> 💡 **这是个好习惯**：**先把你想要的条件用中文/英文说一遍，
> 再逐词翻译成代码。** 直接想代码容易把减法写反。

## §3.1 课件那个函数 + 三个样例

````python
def my_thermo_stat(temp, desired_temp):
    if temp < desired_temp - 5:
        return 'Heat'
    elif temp > desired_temp + 5:
        return 'AC'
    else:
        return 'Off'


print(my_thermo_stat(65, 75))   # Output: Heat
print(my_thermo_stat(75, 65))   # Output: AC
print(my_thermo_stat(65, 63))   # Output: Off
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `def my_thermo_stat(temp, desired_temp):` | 两个参数：当前温度、目标温度 |
| 2 | `if temp < desired_temp - 5:` | ⚠️ **先算右边** `desired_temp - 5`，再比较 |
| 3 | `return 'Heat'` | 太冷了，开暖气 |
| 4 | `elif temp > desired_temp + 5:` | ⚠️ 到这里说明第一个条件不成立（**不低于目标 5 度**），现在问「是不是高于目标 5 度」 |
| 5 | `return 'AC'` | 太热了，开空调 |
| 7 | `return 'Off'` | 其余情况：**在目标温度 ±5 度以内**，关机 |

**三个样例逐个验算：**

| 调用 | 第一个条件 | 第二个条件 | 返回 |
|---|---|---|---|
| `my_thermo_stat(65, 75)` | 65 < 75−5 = **70** -> True | 不看了（命中即停） | `Heat` |
| `my_thermo_stat(75, 65)` | 75 < 65−5 = 60 -> False | 75 > 65+5 = **70** -> True | `AC` |
| `my_thermo_stat(65, 63)` | 65 < 63−5 = 58 -> False | 65 > 63+5 = 68 -> False | `Off` |

**三个输出和课件完全一致。**

> 💡 **注意第三个样例：`my_thermo_stat(65, 63)`**
> 当前 65 度，目标 63 度 —— **当前温度其实比目标高了 2 度**，
> 但因为在 ±5 的死区里，所以返回 `Off`。
>
> **这就是「死区（deadband）」** —— 温控系统故意留的一段「不动作区间」，
> 免得在目标温度附近反复开关（物理上叫「避免振荡」，工程上叫「防抖」）。
> **这个函数比看起来更有物理意义。**

## §3.2 🧪 把整个温度范围扫一遍

课件只给了 3 个样例。我们把 20 个温度全跑一遍，**看清这个函数的「全貌」**。

````python
desired = 70
print(f'目标温度 = {desired}，死区 = 正负 5 度')
print()
print('当前温度   与目标之差   返回')
print('-' * 40)
for temp in range(60, 81, 2):
    result = my_thermo_stat(temp, desired)
    diff = temp - desired
    print(f'{temp:>6}   ({diff:+3d} 度)   ->   {result}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `desired = 70` | 固定目标温度 |
| 6 | `for temp in range(60, 81, 2):` | 从 60 到 80，**每隔 2 度**取一个（共 11 个） |
| 8 | `diff = temp - desired` | 算出「差多少度」 |
| 9 | `{diff:+3d}` | **`+` 号让正数也带正号**，这样正负一目了然 |

**你会看到这样一个「分段」结构：**

| 当前温度 | 与目标之差 | 返回 |
|---|---|---|
| 60, 62 | −10, −8 | **Heat**（低于目标超过 5 度） |
| 64 | −6 | **Heat** |
| **66, 68, 70, 72, 74** | **−4 ~ +4** | **Off**（死区内） |
| **76** | **+6** | **AC** |
| 78, 80 | +8, +10 | **AC** |

> 💡 **这就是一个「分段函数」的图像。** 画出来是一条阶梯：
> 左边一段 Heat、中间一段 Off、右边一段 AC。
>
> ⚠️ **注意边界：差值恰好等于 ±5 时走的是 `Off`。**
> 因为条件是「**more than** 5 degrees」—— 严格大于。
>
> 🧪 **动手改一下**：把 `temp < desired_temp - 5` 改成 `<=`，
> 再跑一遍这个扫描，看 `temp = 65`（差值 −5）的结果从哪里变成哪里。
> **这就是「边界条件」的实测方法。**

## §3.3 🧪 换一种写法：用「差值」当条件

课件的写法是「拿温度跟 目标±5 比」。也可以先算出差值再比。
**两种写法应该给出完全相同的结果** —— 我们来验证这件事。

````python
def thermo_by_diff(temp, desired_temp):
    """先算出差值，再判断 —— 和课件写法等价吗？"""
    diff = temp - desired_temp
    if diff < -5:
        return 'Heat'
    elif diff > 5:
        return 'AC'
    else:
        return 'Off'


print('温度   课件写法   差值写法   一致？')
print('-' * 45)
all_same = True
for temp in range(50, 91):
    a = my_thermo_stat(temp, 70)
    b = thermo_by_diff(temp, 70)
    ok = 'OK' if a == b else '不一致'
    if a != b:
        all_same = False
    if temp % 5 == 0:
        print(f'{temp:>4}   {a:<9} {b:<9} {ok}')
print()
print(f'  在 50~90 这 41 个温度上，两种写法完全一致？ {all_same}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-8 | `thermo_by_diff` | 先 `diff = temp - desired_temp`，再和 ±5 比 |
| 16-18 | 遍历 50~90 **每一个整数** | 41 个温度，逐个对比两个函数 |
| 17-21 | `if a != b:` | 只要发现一个不一致就标记 `all_same = False` |
| 19-20 | `if temp % 5 == 0:` | 只打印能被 5 整除的那些，避免刷屏 |

**输出**：`在 50~90 这 41 个温度上，两种写法完全一致？ True`

> 💡 **这一格演示的是「测试」的基本思想**：
> **不要只试几个样例就下结论，要遍历一个范围。**
>
> 课件给了 3 个样例，我们跑了 41 个 —— **覆盖度完全不同**。
> 3 个样例可能碰巧全对，41 个连续值就很难「碰巧」了。
>
> ⚠️ **但 41 个整数也还不够。** 真正的边界在 `temp = 65` 和 `temp = 75`
> （也就是差值恰好 ±5 的地方），**这两个值必须专门测**。
>
> 📎 这种「自动对比两个实现」的做法，在软件工程里叫**差分测试（differential testing）** ——
> 写新版本时，拿旧版本当标准答案，逐一对拍。**这是最省力的测试方法之一。**

## §3.4 📌 勘误：课件这个例子没有 docstring，也没有单位

课件 C3 的 Best Practices 里明确要求「Comment extensively」，
C4 的课件 P3 给了这个完整函数，**但函数里一行注释都没有**。

更重要的是：**函数没有说明参数的单位。**

````python
print('--- 这个函数的参数是什么单位？看课件看不出来 ---')
print(my_thermo_stat(65, 75), '  <- 65 和 75 是摄氏度？华氏度？开尔文？')
print()
print('同一个函数，换个单位，物理意义完全变了：')
print('  当作摄氏度：65 度到 75 度 —— 合理的室温')
print('  当作华氏度：65 华氏度约 18.3 度，75 华氏度约 23.9 度 —— 也说得通')
print('  当作开尔文：65 K = -208 度 —— 这是液氮温度，不是室温')
print()
print('--- 补上 docstring 之后，这些问题就没了 ---')


def my_thermo_stat_doc(temp, desired_temp):
    """根据当前温度和目标温度，决定恒温器的动作。

    Parameters
    ----------
    temp : float
        当前温度（摄氏度）。
    desired_temp : float
        目标温度（摄氏度）。

    Returns
    -------
    str
        'Heat'（低于目标 5 度以上）/ 'AC'（高于目标 5 度以上）/ 'Off'（死区内）。

    Notes
    -----
    死区为 ±5 度（严格大于 5 才动作），用于防止在目标温度附近反复开关。
    """
    if temp < desired_temp - 5:
        return 'Heat'
    elif temp > desired_temp + 5:
        return 'AC'
    else:
        return 'Off'


print(my_thermo_stat_doc.__doc__)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2-8 | 单位讨论 | 同一组数字 `(65, 75)`，在不同单位下物理意义天差地别 |
| 14-34 | `my_thermo_stat_doc` | **同一个函数，加上完整 docstring** |
| 24-26 | `Returns` 段 | 明确写出三种返回值**各自的条件** —— 这些信息在课件里是散落在 bullet 里的 |
| 27-29 | `Notes` 段 | 解释了「死区」的**设计意图**（防振荡），这是课件完全没提的 |
| 37 | `print(...__doc__)` | 打印出来，证明这些说明真的跟着函数走了 |

> 📌 **勘误说明**：课件 P3 的这个函数**功能完全正确**，但作为教学样本有三个缺憾：
>
> | 缺憾 | 后果 |
> |---|---|
> | 没有 docstring | 违反了它自己 P1 的 Best Practices |
> | 没写单位 | **物理代码里最危险的遗漏** —— 单位错会让结果差几个数量级，而且不报错 |
> | 没解释 ±5 的含义 | 读者不知道这是「防振荡死区」，会以为是随便定的数 |
>
> 💡 **给物理专业的三条硬规矩**：
> 1. **每个物理量的参数，docstring 里必须写单位**
> 2. **每个「魔法数字」（像这里的 5），都要说明它为什么是这个值**
> 3. **每个函数，至少一行 docstring**
>
> 📎 第 1 条在真实项目里能救命：NASA 的火星气候轨道器就是因为
> 一个团队用公制、一个团队用英制，**直接撞毁**（1999 年，损失 3.27 亿美元）。

# §4 · 三元运算符（P4–P5）

## 课件原文（P4）

- **Definition:** Ternary operators, also known as **conditional expressions**, provide a concise way to write simple if-else logic **in a single line of code**.
- **Syntax:**

```python
expression_if_true if condition else expression_if_false
```

```python
is_student = True
person = 'student' if is_student else 'not student'
print(person)   # Output: student
```

- **Benefits:**
  - Simplifies code for basic conditionals.
  - Useful for making code more concise and readable.
  - Often employed in **list comprehensions** and similar constructs.

## 课件原文（P5）

**Equivalent Block for Ternary Operator**

```python
is_student = True
if is_student:
    person = 'student'
else:
    person = 'not student'
print(person)   # Output: student
```

- **Comparison:**
  - The ternary operator offers a **more compact representation**, reducing the number of lines and increasing readability for **simple** conditionals.
  - The traditional if-else block is better suited for **more complex** conditional logic.

## 人话版

**三元运算符 = 「一行写完的 if-else」。**

**⚠️ 但它的语序和 if-else 完全相反，第一次看一定会懵。**

| 写法 | 语序 |
|---|---|
| `if-else` 语句 | 「**如果**……**那么**……**否则**……」 |
| 三元运算符 | 「**那么**……**如果**……**否则**……」 |

对照着看：

```python
# if-else 语句（5 行）
if is_student:
    person = 'student'
else:
    person = 'not student'

# 三元运算符（1 行）—— 注意语序
person = 'student' if is_student else 'not student'
#        ^^^^^^^^^ 是「真的要用的值」，放最前面
#                     ^^^^^^^^^^ 条件在中间
#                                  ^^^^^^^^^^^^^ 条件不成立时的值在最后
```

> 💡 **怎么读这一行？** 建议**从中间读起**：
> 「`if is_student`（如果他是学生）→ 取前面那个 `'student'`；
> `else` → 取后面那个 `'not student'`。」
>
> **记忆口诀：「要什么，看条件，否则另一样」。**

**⚠️ 什么时候绝对不要用三元运算符**

课件 P5 说得很清楚：**simple 才用，complex 就不该用。**

```python
# 千万别这样写（四层嵌套）
grade = 'A' if s >= 90 else 'B' if s >= 80 else 'C' if s >= 70 else 'F'

# 该用 if-elif-else
if s >= 90:
    grade = 'A'
elif s >= 80:
    grade = 'B'
elif s >= 70:
    grade = 'C'
else:
    grade = 'F'
```

**判据**：**嵌套的三元运算符，一律改成 `if-elif`。**

## §4.1 课件两个版本对照

````python
print('=== 版本 1：三元运算符（课件 P4）===')
is_student = True
person = 'student' if is_student else 'not student'
print('  person =', person)

print()
print('=== 版本 2：if-else 语句（课件 P5）===')
is_student = True
if is_student:
    person = 'student'
else:
    person = 'not student'
print('  person =', person)
print()
print('  两个版本结果一样，区别只在「行数」')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 3 | `person = 'student' if is_student else 'not student'` | **一行搞定**。读法：`if is_student` 成立 -> 取 `'student'`；否则取 `'not student'` |
| 8-12 | 同样的 if-else 语句 | **5 行**，但语序和自然语言一致 |
| 13 | `print(person)` | 两版结果都是 `student`（和课件一致） |

> 💡 **三元运算符返回的是一个「值」** —— 这是它和 if 语句最本质的区别。
>
> | | 三元运算符 | if-else 语句 |
> |---|---|---|
> | 是什么 | **表达式**（有值） | **语句**（没有值） |
> | 能放在 | 赋值右边、函数参数、列表推导式里 | 只能单独成行 |
> | 能写几行逻辑 | **只能一个表达式** | 想写多少写多少 |
>
> ⚠️ **「表达式 vs 语句」这个区别非常重要**，
> 因为**只有表达式才能塞进别的地方**。比如：
>
> ```python
> print('成年' if age >= 18 else '未成年')            # 塞进函数参数
> labels = ['奇' if x % 2 else '偶' for x in nums]   # 塞进推导式（C5）
> ```
>
> **这两种用法是 if 语句做不到的。** 这才是三元运算符存在的真正理由。

## §4.2 🧪 三元运算符能塞进去的地方

````python
age = 20

print('--- 用法 1：直接塞进函数参数 ---')
print('  ' + ('成年' if age >= 18 else '未成年'))
print()

print('--- 用法 2：作为函数的返回值 ---')


def label_age(a):
    return '成年' if a >= 18 else '未成年'


for a in [10, 17, 18, 30]:
    print(f'  {a:>3} 岁 -> {label_age(a)}')
print()

print('--- 用法 3：塞进 f-string ---')
score = 85
print(f'  分数 {score}：{"及格" if score >= 60 else "不及格"}')
print()

print('--- 用法 4：塞进列表推导式（C5 会细讲）---')
nums = [1, 2, 3, 4, 5, 6]
parity = ['奇' if x % 2 else '偶' for x in nums]
print('  nums   =', nums)
print('  parity =', parity)
print()

print('--- 用法 5：给数据「设上限」---')
raw = [3, 12, 7, 25, 1]
capped = [x if x <= 10 else 10 for x in raw]
print('  raw    =', raw)
print('  capped =', capped, '  <- 超过 10 的一律压到 10')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4 | `print('  ' + ('成年' if ... else '未成年'))` | ⚠️ **括号是必需的**，否则 `+` 的优先级会和三元运算符打架 |
| 8-9 | `return '成年' if a >= 18 else '未成年'` | **最常见的用法**：函数根据条件返回两个不同值 |
| 15 | `f'... {"及格" if score >= 60 else "不及格"}'` | 塞进 f-string 的大括号里。⚠️ **里面的字符串用双引号**，避免和外层的单引号打架 |
| 22 | `['奇' if x % 2 else '偶' for x in nums]` | **塞进推导式**。这正是课件说的 "Often employed in list comprehensions" |
| 28 | `[x if x <= 10 else 10 for x in raw]` | **给数据设上限**（物理里叫「限幅 / clipping」） |

> 💡 **第 28 行那个「限幅」操作在物理实验里非常实用**：
> 传感器偶尔会给出离谱的野值，直接扔掉可能不合适，压到上限往往是安全的做法。
>
> ⚠️ **注意第 4 行那两个括号**。不加括号会怎样？
> `print('  ' + '成年' if age >= 18 else '未成年')` ——
> 这会变成「先算 `'  ' + '成年'`，再用三元判断」，**结果碰巧一样，但完全是另一套逻辑**。
> **三元运算符的优先级很低，混用时一定加括号。**

## §4.3 ⚠️ 什么时候不要用三元运算符

课件 P5 的结论，用一个实测来说明。

````python
print('=== 可以接受：一层三元 ===')


def grade_simple(s):
    return '及格' if s >= 60 else '不及格'


print(' ', [grade_simple(s) for s in [95, 60, 59, 30]])

print()
print('=== 勉强能用：两层（已经开始难读了）===')


def grade_two(s):
    return 'A' if s >= 90 else ('B' if s >= 60 else 'F')


print(' ', [grade_two(s) for s in [95, 75, 30]])
print('  你花了几秒才看懂这一行？')

print()
print('=== 不该用：四层（噩梦）===')


def grade_bad(s):
    return 'A' if s >= 90 else 'B' if s >= 80 else 'C' if s >= 70 else 'D' if s >= 60 else 'F'


print(' ', [grade_bad(s) for s in [95, 85, 75, 65, 30]])
print('  虽然结果是对的，但这一行没法维护、没法调试')

print()
print('=== 正确做法：超过一层就用 if-elif-else ===')


def grade_good(s):
    if s >= 90:
        return 'A'
    elif s >= 80:
        return 'B'
    elif s >= 70:
        return 'C'
    elif s >= 60:
        return 'D'
    else:
        return 'F'


print(' ', [grade_good(s) for s in [95, 85, 75, 65, 30]])
print('  行数多了，但一眼就能看懂每一档')
print()
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-5 | `grade_simple` | 一层三元：**读起来毫无障碍** |
| 12-13 | `grade_two` | 两层三元：**必须加括号**，而且要来回看好几遍 |
| 20-21 | `grade_bad` | 四层三元：**一坨**。虽然能跑、结果也对，但改一个数就得重新理解整行 |
| 27-38 | `grade_good` | 老老实实的 `if-elif-else`：**行数多了，但每一档清清楚楚** |

**四个函数的输出完全一样**（A/B/C/D/F 分级），差别只在**可读性**。

> ⚠️ **为什么「能跑」不够？**
> 因为代码是**给人看的**，只是顺便给机器跑。
> 一行 90 个字符的四层三元，三个月后你自己都看不懂。
>
> 💡 **实用判据（照这个来就不会错）**：
>
> | 情况 | 用什么 |
> |---|---|
> | 只有两个分支、各自是一个**短表达式** | 三元运算符 |
> | 超过两个分支 | `if-elif-else` |
> | 分支里的表达式很长 | `if-elif-else` |
> | 要放进推导式、函数参数、f-string | **只能**用三元运算符 |
>
> 📌 **课件 P5 说得完全正确**：「simple conditionals 用三元，complex conditional logic 用传统 if-else」。
> 这一节只是把「怎么算 simple」量化成了上面的判据。

| `else` | 可选；`if` 和 `else` **二选一，必然执行一个** |
| `elif` | 「否则如果」；**可以有多个 `elif`，只能有一个 `else`** |
| 命中即停 | 一旦某个条件为真，**后面的条件根本不看** -> **顺序会影响结果** |
| 缩进 | **4 个空格**；缩进结束 = 块结束；顶格的行不属于任何分支 |
| 三元运算符 | `值1 if 条件 else 值2` —— **语序和 if-else 相反**，只在简单时用 |

## 三种分支结构，怎么选

| 情况 | 用什么 | 例子 |
|---|---|---|
| 只有两个结果 | `if / else` | 及格 / 不及格 |
| 多个互斥的结果 | `if / elif / else` | 成绩 A~F |
| 多个**可以同时成立**的判断 | 独立的 `if` | 「优秀」+「及格」两个标签 |
| 简单二选一，且要**放进别的地方** | 三元运算符 | `f'... {"及格" if s>=60 else "不及格"}'` |

## 三个坑（本章全部实测过）

1. **边界值走哪条路？** `temp > 30` 不包含 30。**写完 `if` 一定要喂边界值跑一遍。**
2. **`elif` 顺序写反** -> 宽的条件在前面，「窄」的分支永远轮不到，**而且不报错，只是静默给错答案**。
3. **三元运算符没有括号** -> 和 `+`、`and` 混用时优先级会咬人。

## 三个好习惯

1. **先写「人话条件」，再翻译成代码。** 「当前温度比目标低 5 度以上」-> `temp < desired_temp - 5`。
   直接想代码容易把减法的方向写反。
2. **缩进用 Tab 键，不要在空格和 Tab 之间来回切。** 混用会报 `TabError`，而且报错信息很难懂。
3. **超过一层的三元运算符，一律改成 `if-elif-else`。**
