## Slide 1 — Complexity and Big-O Notation

- **Complexity** describes how the difficulty or resource use of a function changes with input size, often written as *n*.
- **Basic operations** include additions, multiplications, and function calls. Counting them approximates runtime.
- **Big-O notation** describes the growth rate for large inputs. It keeps the dominant term and ignores constants and lower-order terms.
- Example: `4n² + n + 1` has growth `O(n²)`.

## Slide 2 — Examples of Algorithm Complexity

### Quadratic time: `O(n²)`

The nested loops perform *n × n* iterations.

```python
def quadratic_work(n):
    total = 0
    for i in range(n):
        for j in range(n):
            total += i * j
    return total

quadratic_work(4)
```

There are `n²` inner-loop executions, so the runtime is `O(n²)`.

### Linear time: `O(n)`

This iterative Fibonacci example appends one value per loop iteration.

```python
def my_fib_iter(n):
    out = [1, 1]
    for i in range(2, n):
        out.append(out[i - 1] + out[i - 2])
    return out

my_fib_iter(8)
```

The loop runs about *n* times, so the runtime is `O(n)` (and the list uses `O(n)` space).

## Slide 3 — Recursive and Logarithmic Complexity

### Exponential time: `O(2ⁿ)`

The recursive Fibonacci function branches into two calls for most values of *n*.

```python
def my_fib_rec(n):
    if n < 2:
        return 1
    return my_fib_rec(n - 1) + my_fib_rec(n - 2)

my_fib_rec(8)
```

The number of calls grows exponentially (more precisely, this simple recursive Fibonacci algorithm is `O(φⁿ)`, commonly bounded by `O(2ⁿ)`).

### Logarithmic time: `O(log n)`

Halving the input each iteration takes a logarithmic number of steps.

```python
def my_divide_by_two(n):
    steps = 0
    while n > 1:
        n /= 2
        steps += 1
    return steps

my_divide_by_two(64)
```

## Slide 4 — Why Complexity Matters

Complexity describes how runtime or space requirements grow with input size.

| Growth | Example | Typical behavior |
|---|---|---|
| Logarithmic | `O(log n)` | Input is repeatedly reduced, often by half |
| Linear | `O(n)` | Work grows in proportion to input size |
| Quadratic | `O(n²)` | Often comes from two nested loops |
| Exponential | `O(2ⁿ)` | Quickly becomes impractical as input grows |

The slide’s graph compares logarithmic, linear, quadratic, and exponential growth. Exponential algorithms become infeasible quickly; polynomial algorithms are generally more manageable as input grows.

## Slide 5 — Algorithm Scaling and Performance

The slide compares exponential algorithms such as `O(2ⁿ)` with polynomial algorithms such as `O(nᶜ)`.

- Doubling compute power increases the solvable input size of an exponential-time problem by only about one unit.
- For a quadratic-time problem, doubling the available runtime permits an input-size increase by a factor of about `√2` (assuming the same machine and implementation).
- The iterative Fibonacci algorithm scales far better than the simple recursive version, so iterative or otherwise optimized methods are preferred for performance-critical work.

## Slide 7 — Profiling in Python

**Profiling** measures code runtime and resource use to locate bottlenecks. It helps focus optimization on the portions of a program that consume the most time or resources.

Tools introduced in the lecture:

- `%time` and `%timeit`: Jupyter/IPython timing magics.
- `cProfile`: Python’s built-in function-level profiler.
- `line_profiler`: optional package for measuring individual lines.

## Slide 8 — Magic Commands for Time Measurement

### Single line

`%time` measures one statement. The PDF shows `%time sum(range(200))`; its sample timing is hardware- and session-dependent.

### Whole cell

`%%time` measures the full cell. Run this cell in Jupyter/IPython:

```python
%%time
s = 0
for i in range(200):
    s += i
```

### Repeated measurement

`%timeit` runs a statement repeatedly and reports a typical time. The sample below is approximate and will vary by computer:

```python
%timeit sum(range(200))
```

## Slide 9 — Line Profiling

Line profiling measures time spent on individual lines. The example below follows the slide: it creates a random array, then adds the elements one by one. Install `line_profiler` in the notebook environment before running the profiling magic.

```python
import numpy as np

def slow_sum(n, m):
    for i in range(n):
        # Create an array of m random numbers.
        a = np.random.rand(m)
        s = 0
        # Add the array elements one by one.
        for j in range(m):
            s += a[j]
    return s
```

```python
%load_ext line_profiler
%lprun -f slow_sum slow_sum(1000, 10000)
```

The slide’s line-profile output shows the nested loop and per-element addition account for most of the runtime. Actual line numbers and timings depend on the exact code and machine.
