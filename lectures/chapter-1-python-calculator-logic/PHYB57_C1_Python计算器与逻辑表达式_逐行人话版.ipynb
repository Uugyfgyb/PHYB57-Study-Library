# §1 · 把 Python 当计算器（P1）

## 课件原文（P1）

**Overview**

- Python can be used as a powerful calculator to perform basic arithmetic operations similar to a standard graphing calculator.
- Operations include addition (+), subtraction (-), multiplication (*), division (/), and exponentiation (**).

**Order of Operations**

- Python follows the standard mathematical order of operations: Parentheses, Exponents, Multiplication and Division (from left to right), Addition and Subtraction (from left to right).
- Parentheses can be used to change the order of execution.

**Example Calculation**

```python
(3*4) / (2**2 + 4/2)   # Output: 2.0
```

## 人话版

**七个算术运算符**

| 运算符 | 名字 | 例子 | 结果 | 备注 |
|---|---|---|---|---|
| `+` | 加 | `2 + 3` | `5` | 也能拼字符串、拼列表 |
| `-` | 减 | `7 - 10` | `-3` | 还能当负号：`-5` |
| `*` | 乘 | `6 * 7` | `42` | 字符串 `'ab' * 3` → `'ababab'` |
| `/` | 除 | `7 / 2` | `3.5` | **永远返回 float**，哪怕除得尽 |
| `**` | 乘方 | `2 ** 10` | `1024` | 注意是**两个星号**，不是 `^` |
| `//` | 整除 | `7 // 2` | `3` | 向下取整（不是四舍五入） |
| `%` | 取余 | `7 % 2` | `1` | 判断奇偶、周期性问题的常客 |

**运算优先级**（从高到低，同一级从左往右）

```
1. ( )        括号
2. **         乘方
3. -x  +x     一元正负号
4. *  /  //  %  乘、除、整除、取余
5. +  -       加、减
```

课件那句话的意思就是：**Python 和你中学学的运算顺序一模一样**，不需要重新学一套。

> 💡 **一句话抓住本质**：括号是唯一「想改变顺序就能改变顺序」的工具。**不确定就加括号** ——
> 括号不会让代码变慢，只会让它变清楚。

## §1.1 把课件那一行算一遍

课件只给了一行代码，我们就先原样跑一遍，再**把每一步拆开**，看清楚它为什么等于 2.0。

````python
# 课件 P1 的唯一一行代码，原样抄下来
print((3*4) / (2**2 + 4/2))   # Output: 2.0
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2 | `(3*4) / (2**2 + 4/2)` | 完全照抄课件。括号先算：`(3*4)`=12、`(2**2 + 4/2)`=4+2=6；再相除 12/6 = **2.0** |

> ⚠️ 注意输出是 `2.0` 而不是 `2`。
> **只要用了 `/`，结果就一定是浮点数**，哪怕它正好是个整数。这是 Python 3 的硬规定，
> 和 Python 2 不一样（Python 2 里 `7/2` 会等于 `3`，无数老代码的 bug 都源于此）。

## §1.2 🧪 把这一行拆成四步

上面那一行有 4 个运算挤在一起。把它们一个个拆出来单独跑，就能看清**每一步各自做了什么**。

````python
a = 3 * 4        # 第一个括号
b = 2 ** 2       # 括号里的乘方
c = 4 / 2        # 括号里的除法
d = b + c        # 括号里的加法

print('a = 3*4   =', a)
print('b = 2**2  =', b)
print('c = 4/2   =', c)
print('d = b + c =', d)
print('a / d     =', a / d)   # 最后一步：括号外面那个除法
print()
print('每一步的类型:', type(a).__name__, type(b).__name__, type(c).__name__, type(d).__name__)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `a = 3 * 4` | `=` 是**赋值**，不是数学上的「等于」。这句话读作「把右边算出来的东西，起名叫 a」 |
| 2 | `b = 2 ** 2` | `**` 是乘方。`2**2` 就是 2² = 4 |
| 3 | `c = 4 / 2` | 真除法。结果是 `2.0` —— **带小数点**，这是关键 |
| 4 | `d = b + c` | `4 + 2.0`。int 和 float 相加，**结果自动升级成 float** → `6.0` |
| 6-9 | `print('...', x)` | `print` 可以接好几样东西，用逗号隔开，输出时自动用空格连接 |
| 11 | `type(x).__name__` | `type(x)` 给出类型对象，`.__name__` 取出它的名字字符串（`'int'` / `'float'`） |

**这张表最重要的信息在最后一行**：`a`、`b` 是 `int`，`c`、`d` 是 `float`。
整个式子的「浮点性」是从 `4/2` 那一处传染过来的。

> 💡 **记住这条传染规则**：`int` 和 `float` 混合运算 → 结果一定是 `float`。
> 反过来，只有 `+ - * ** //` 且两边都是 `int` 时，结果才保持 `int`。

## §1.3 🧪 优先级到底怎么生效的

课件说「按标准数学顺序」，那我们就把顺序**故意搞乱**，看 Python 会不会纠正过来。
下面每一组都是「不写括号」和「写括号」的对照。

````python
print('2 + 3 * 4      =', 2 + 3 * 4)        # 乘法先
print('(2 + 3) * 4    =', (2 + 3) * 4)      # 括号改变一切
print()
print('2 ** 3 ** 2    =', 2 ** 3 ** 2)      # 乘方是「右结合」：2**(3**2) = 2**9
print('(2 ** 3) ** 2  =', (2 ** 3) ** 2)    # 强行左结合
print()
print('7 / 2          =', 7 / 2)            # 真除法 → float
print('7 // 2         =', 7 // 2)           # 整除 → 向下取整
print('7 % 2          =', 7 % 2)            # 余数
print('-7 // 2        =', -7 // 2)          # 向下取整 ≠ 向零取整！
print('-7 % 2         =', -7 % 2)
print()
print('10 - 3 - 2     =', 10 - 3 - 2)       # 同级从左往右
print('10 - (3 - 2)   =', 10 - (3 - 2))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `2 + 3 * 4` | 先乘后加 → `14`。Python 不会「从左往右」傻算成 20 |
| 2 | `(2 + 3) * 4` | 括号优先级最高 → `20` |
| 4 | `2 ** 3 ** 2` | **乘方是右结合**：先算右边 `3**2=9`，再算 `2**9=512`。不是 `(2**3)**2=64` |
| 7 | `7 / 2` | 真除法，永远给 float → `3.5` |
| 8 | `7 // 2` | 整除，**向下取整** → `3` |
| 9 | `7 % 2` | 取余 → `1` |
| 10 | `-7 // 2` | ⚠️ 向下取整，所以是 `-4` 而**不是** `-3`。C 语言和 MATLAB 在这里跟 Python 不一样 |
| 11 | `-7 % 2` | 配合上一行，余数取 `1`（因为 `-4*2 + 1 = -7`）。**Python 保证 `a == (a//b)*b + a%b`** |
| 13 | `10 - 3 - 2` | 同级从左往右 → `5` |
| 14 | `10 - (3 - 2)` | 加了括号 → `9` |

> ⚠️ **坑 1**：乘方是**右结合**的。`2**3**2` = 512，不是 64。不确定就加括号。
>
> ⚠️ **坑 2**：`//` 是**向下取整**（floor），不是**向零取整**（truncate）。
> 对正数两者一样，对负数就差了 1。
>
> 🧪 顺便验证一条恒等式：`(a // b) * b + (a % b) == a` 对任何整数都成立 —— 这是 `//` 和 `%` 的定义。

## §1.4 🧪 浮点数：计算器的第一个裂缝

课件的标题里写着「similar to a standard graphing calculator」。
计算器有个众所周知的毛病：**0.1 + 0.2 不等于 0.3**。Python 也一样。

````python
print('0.1 + 0.2      =', 0.1 + 0.2)
print('0.1 + 0.2 == 0.3 ->', 0.1 + 0.2 == 0.3)
print()
print('3 * 0.1        =', 3 * 0.1)
print('0.3            =', 0.3)
print('3*0.1 == 0.3   ->', 3 * 0.1 == 0.3)
print()
import math
print('math.isclose(0.1+0.2, 0.3) ->', math.isclose(0.1 + 0.2, 0.3))   # 正确的比较方式
print()
print('--- 用 17 位小数，把「藏起来的差异」逼出来 ---')
print(f'0.1 + 0.2 的 17 位小数 = {0.1 + 0.2:.17f}')
print(f'0.3       的 17 位小数 = {0.3:.17f}')
print(f'1 / 3     的 17 位小数 = {1/3:.17f}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `0.1 + 0.2` | 心里想的是 `0.3`，屏幕上打出 `0.30000000000000004` |
| 2 | `... == 0.3` | 所以这个比较是 **False**。`==` 是「一模一样」的意思，差一个 bit 都算不等 |
| 4-7 | `3 * 0.1` 与 `0.3` | `3 * 0.1` 也等于 `0.30000000000000004`，所以和 `0.3` 比较同样是 `False`。**换一种算法，误差就换一种露法** |
| 11-14 | `f'{x:.17f}'` | 用 **17 位小数**把两个数印出来，差异就藏不住了：`0.1+0.2` 是 `0.30000000000000004`，而 `0.3` 其实是 `0.29999999999999999` |
| 9 | `math.isclose(a, b)` | 浮点数比较的**正确姿势**：问「它俩是不是足够接近」，而不是「是不是一模一样」 |

> ⚠️ **这是本章最值钱的一格。** 记住三句话：
>
> 1. 计算机里没有「实数」，只有「离它最近的那个二进制分数」
> 2. 所以 **永远不要用 `==` 比较两个算出来的浮点数**
> 3. 要比较就用 `math.isclose(a, b)`（或指定容差）
>
> 📎 这个现象的原因（IEEE 754 二进制浮点表示）**整章都在 `PHYB57_C9` 里讲**，
> 这里只是先让你看一眼它的脸。

## §2.1 把课件那两行跑一遍

````python
import math

print('math.pi            =', math.pi)
print('math.sin(math.pi/2)=', math.sin(math.pi / 2))     # Output: 1.0
print('math.exp(math.log(10)) =', math.exp(math.log(10))) # Output: 10.000000000000002
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `import math` | 拉开 `math` 这个抽屉。**每次开新 notebook 都要重新 import** —— 每个 notebook 是一个独立的进程 |
| 3 | `math.pi` | 取 π 的值（注意没有括号） |
| 4 | `math.sin(math.pi / 2)` | 先算 `math.pi/2`（≈1.5708），再取正弦 → `1.0`。**正好等于 1.0，一点误差都没有** |
| 5 | `math.exp(math.log(10))` | 先 `ln 10`（≈2.302585），再 `e^那个数`。数学上应该回到 10 |

**看第 5 行的输出：`10.000000000000002`。** 课件把它原样印在幻灯片上了 —— 这不是排版错误，
它就是真实结果。

> ⚠️ **为什么 sin(π/2) 完美，exp(log(10)) 却差了一点？**
>
> 因为 π/2 在二进制里存的是个近似值，但 `sin` 在 π/2 附近**非常平坦**（导数为 0），
> 输入的小误差传到输出就几乎没了；而 `exp`/`log` 是一对**互为逆运算**的函数，
> 两次舍入误差不会互相抵消，反而会累加暴露出来。
>
> 📌 **勘误 1**：课件把 `10.000000000000002` 直接当成正常输出展示，没说这是浮点误差。
> 这容易让人以为「Python 算错了」。**它没算错，它是在用有限位数逼近。**
> 正确的心态：只要看到一堆 `999999` 或 `000002`，就知道该上 `math.isclose` 了。

## §2.2 🧪 把 math 抽屉倒出来看看

课件只列了 5 个成员，但 `math` 里其实有 50 多个。下面挑物理课上最常用的跑一遍。

````python
print('--- 幂与根 ---')
print('math.sqrt(2)            =', math.sqrt(2))          # 平方根
print('math.pow(2, 10)         =', math.pow(2, 10))       # 等价于 2**10，但返回 float
print('math.hypot(3, 4)        =', math.hypot(3, 4))      # 直角三角形斜边，比 sqrt(a²+b²) 更稳

print()
print('--- 指数与对数 ---')
print('math.exp(1)             =', math.exp(1))           # e¹
print('math.log(math.e)        =', math.log(math.e))      # 自然对数，底是 e
print('math.log(100, 10)       =', math.log(100, 10))     # 第二个参数 = 底数 → log₁₀
print('math.log10(1000)        =', math.log10(1000))
print('math.log2(1024)         =', math.log2(1024))

print()
print('--- 三角与反三角 ---')
print('math.sin(math.pi/6)     =', math.sin(math.pi/6))   # 应该 = 0.5
print('math.cos(math.pi)       =', math.cos(math.pi))     # 应该 = -1
print('math.atan2(1, 1)        =', math.atan2(1, 1))      # 比 atan(y/x) 好：能分辨象限
print('math.degrees(math.pi)   =', math.degrees(math.pi)) # 弧度 → 角度
print('math.radians(180)       =', math.radians(180))     # 角度 → 弧度

print()
print('--- 取整与组合 ---')
print('math.floor(2.7)         =', math.floor(2.7))
print('math.ceil(2.1)          =', math.ceil(2.1))
print('math.factorial(5)       =', math.factorial(5))     # 5! = 120（课件 Tips 里点名的函数）
print('math.gcd(24, 36)        =', math.gcd(24, 36))      # 最大公约数
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 3 | `math.sqrt(2)` | √2 ≈ 1.4142135623730951。**参数不能是负数**，否则 `ValueError` |
| 4 | `math.pow(2, 10)` | 和 `2**10` 算出来一样，但**一定返回 float**（`1024.0`）。超大数据时 `**` 更准 |
| 5 | `math.hypot(3, 4)` | 直角边 3、4 的斜边 = 5.0。**比手写 `sqrt(3**2+4**2)` 好**：它内部做了防溢出处理 |

## §3.3 🧪 复数的常用操作

物理里复数不是花架子。这一格把它的四大件都用一遍。

````python
z = 3 + 4j

print('z            =', z)
print('实部 z.real  =', z.real)
print('虚部 z.imag  =', z.imag)
print('共轭 z.conjugate() =', z.conjugate())
print('模   abs(z)  =', abs(z))              # 3-4-5 直角三角形
print()
print('虚数单位平方 1j**2 =', 1j**2)          # 应该 = -1
print('(2+5j) + (1-3j)   =', (2 + 5j) + (1 - 3j))
print('(2+5j) * (1-3j)   =', (2 + 5j) * (1 - 3j))   # 复数乘法：模相乘、角相加
````

| 6 | `abs(z)` | 复数的模 |z| = √(3²+4²) = `5.0`。**`abs()` 对复数就是取模**，不是「去掉符号」 |
| 8 | `1j**2` | 定义式 i² = −1 → `(-1+0j)` |
| 9 | 复数加法 | 实部加实部、虚部加虚部 → `(3+2j)` |

## §3.4 🧪 欧拉公式：复数 + math 模块合体

物理里最有名的一条式子 e^{iθ} = cos θ + i sin θ，用刚学的两样东西一行就能验证。

````python
theta = math.pi / 3        # 60°

left = math.e ** (1j * theta)     # 左边：e^{iθ}
right = complex(math.cos(theta), math.sin(theta))   # 右边：cosθ + i·sinθ

print('e^{i·π/3}        =', left)
print('cos(π/3)+i·sin(π/3) =', right)
print('两者相等？        ', math.isclose(left.real, right.real) and math.isclose(left.imag, right.imag))
print()
print('特例 θ = π：e^{iπ} =', math.e ** (1j * math.pi))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `theta = math.pi / 3` | 60° 的弧度值 |
| 3 | `math.e ** (1j * theta)` | 指数位置上放个**复数**，Python 的 `**` 会自动走复数运算。**复数指数不需要额外装库** |
| 4 | `complex(cos, sin)` | 把实部和虚部拼成一个复数。这里正好演示了「`complex()` 在什么时候比 `a+bj` 好用」 |
| 6 | 比较 | ⚠️ 这里**没有**写 `left == right`，而是分别比实部和虚部的 `isclose`。**这就是 §1.4 那条规矩的实战用法** |
| 9 | `math.e ** (1j * math.pi)` | 欧拉恒等式 e^{iπ} = −1 |

**看第 9 行的输出**：是 `(-1+1.2246467991473532e-16j)`，**不是**干干净净的 `(-1+0j)`。

> ⚠️ 这个 `1.22e-16` 就是浮点误差，它的大小和 `math.pi` 本身的舍入误差同量级。
> 数学上 e^{iπ} = −1 严格成立；计算机上它等于「−1 加上一个 10⁻¹⁶ 量级的小尾巴」。
>
> 💡 **物理人的心法**：看到 `1e-16` 这种量级，先问自己「这是物理效应还是数值噪声」。
> 在这里显然是噪声，因为它是机器精度的量级。**这就是 `PHYB57_C9` 要教你的判断力。**

# §6 · 逻辑表达式与比较运算符（P10）

## 课件原文（P10）

**Logical Expressions**

- A logical expression evaluates to either `True` or `False`.
- Python uses the **boolean** data type for logical expressions.
- `True` is equivalent to 1
- `False` is equivalent to 0
- **Example:** Expression: `3 < 4` → Python evaluates it as `True`.

**Comparison Operators**

| 运算符 | 含义 |
|---|---|
| `>` | Greater than |
| `>=` | Greater than or equal to |
| `<` | Less than |
| `<=` | Less than or equal to |
| `!=` | Not equal |
| `==` | Equal |

**Example of Comparison:** Given `a = 4` and `b = 2`:

- `a > b` evaluates to `True`
- `a == b` evaluates to `False`

## 人话版

**「逻辑表达式」= 一个能回答是/非的问题。**

`3 < 4` 不是一个「数」，它是一个**判断**，答案是「真」。Python 把「真」写成 `True`，
「假」写成 `False` —— 首字母**必须大写**。

**六个比较运算符**

| 运算符 | 读作 | 例子 | 结果 |
|---|---|---|---|
| `>` | 大于 | `4 > 2` | `True` |
| `>=` | 大于等于 | `4 >= 4` | `True` |
| `<` | 小于 | `4 < 2` | `False` |
| `<=` | 小于等于 | `4 <= 2` | `False` |
| `!=` | 不等于 | `4 != 2` | `True` |
| `==` | **等于**（两个等号！） | `4 == 2` | `False` |

> ⚠️ **这是初学者第一大坑：`=` 和 `==` 是两个完全不同的东西。**
>
> | 写法 | 名字 | 干什么 |
> |---|---|---|
> | `x = 4` | **赋值** | 把 4 存进 x，**这个式子本身不产生 True/False** |
> | `x == 4` | **比较** | 问「x 是不是 4」，产生 `True` 或 `False` |
>
> 在 `if` 里把 `==` 写成 `=` 是最经典的语法错误。

**`True` = 1，`False` = 0**

课件专门用两条 bullet 说了这件事，因为它真的能用：

## §6.1 课件的例子 + True/False 的数值身份

````python
a = 4
b = 2

print('a > b   ->', a > b)     # True
print('a == b  ->', a == b)    # False
print()
print('True  == 1  ->', True == 1)
print('False == 0  ->', False == 0)
print('True + True ->', True + True)          # 布尔值可以直接参与算术！
print('True * 5    ->', True * 5)
print()
print('bool 是 int 的子类？', issubclass(bool, int))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-2 | `a = 4` / `b = 2` | 赋值。**这是 `=`，不产生 True/False** |
| 4-5 | `a > b` / `a == b` | 比较。**这是产生 True/False 的地方**，直接对着课件的例子 |
| 7-8 | `True == 1` | 课件说「True is equivalent to 1」—— 实测就是 `True` 本身（而不是「约等于」） |
| 9-10 | `True + True` | 布尔值能当数字用：`1 + 1 = 2`、`1 * 5 = 5` |
| 12 | `issubclass(bool, int)` | 从类型系统上确认：`bool` **就是** `int` 的一个特例 |

> 💡 **为什么要知道 `True == 1`？** 因为它带来一个非常好用的技巧：**用 `sum()` 数「满足条件的个数」**。
> 下面这一格就演示。

## §6.2 🧪 用 sum() 数数：True 当 1 用的实战

物理实验里最常见的一句话是「这批数据里有几个符合条件」。有了 `True == 1`，一行就能数出来。

````python
data = [3, 7, 1, 9, 4, 8, 2, 6, 5, 10]

n_big = sum(x > 5 for x in data)      # 数「大于 5 的有几个」
print('大于 5 的数据有', n_big, '个')
print('占比', n_big / len(data))
print()
print('最大值 =', max(data), '  最小值 =', min(data), '  平均 =', sum(data) / len(data))
print()
# 对照写法：用显式的 if 累加
count = 0
for x in data:
    if x > 5:
        count += 1
print('用 for + if 数出来也是', count, '个  —— 和上面那一行完全等价')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `data = [...]` | 10 个数据点 |
| 3 | `sum(x > 5 for x in data)` | 对每个 x 算 `x > 5`，得到 10 个 `True`/`False`；`sum` 把它们当 1/0 加起来 —— 和就是「满足条件的个数」 |
| 8-11 | 对照写法 | 用 `for` + `if` 得到的答案一模一样，但**长了 5 行** |
| 12 | `print(...)` | 两个答案对照，证明第 3 行确实在数数 |

> ⚠️ 这里出现了两个还没正式讲的东西：`for x in data` 和 `if x > 5`。
> 它们分别在 **C5（循环）** 和 **C4（条件判断）** 里详细讲。这里先用一下，
> **重点体会的是「True 能当 1 加」这件事**。

## §6.3 🧪 链式比较：Python 独有的写法

数学里写 0 < x < 10 很自然。很多语言里这会被解释成 `(0 < x) < 10`，从而永远为真。
**Python 直接支持数学写法**，而且语义和数学完全一致。

````python
x = 5
print('0 < x < 10        ->', 0 < x < 10)          # 数学写法，Python 原生支持
print('(0 < x) < 10      ->', (0 < x) < 10)        # 别的语言的写法：True < 10 → True < 1？
print('0 < x and x < 10  ->', 0 < x and x < 10)    # 等价展开
print()
x = 15
print('x = 15 时 0 < x < 10 ->', 0 < x < 10)
print('x = 15 时 0 < x < 10 < 20 ->', 0 < x < 10 < 20)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2 | `0 < x < 10` | Python 把 `a < b < c` 理解成 `(a < b) and (b < c)`，**b 只会被求值一次** |
| 3 | `(0 < x) < 10` | 加了括号就退化成「先算 0<5 → True，再问 True < 10」。因为 `True == 1`，所以结果是 `True` —— **碰巧对了，但逻辑是错的** |
| 4 | `0 < x and x < 10` | 完全等价的手写版。**建议：哪怕用链式写法，心里也要能翻译成这一行** |
| 7-8 | x = 15 | 链式比较任何一环断了，整体就是 False |

> ⚠️ **坑**：`(0 < x) < 10` 在 x=5 时碰巧是真，在 x=15 时也碰巧是真（False < 10 → 0 < 10 → True）。
> **它永远是真。** 这种 bug 不会报错，只会让你的判断条件静默失效 —— 这是最难查的一类 bug。
>
> 💡 **好习惯**：范围判断一律写 `0 < x < 10`，不要写 `(0 < x) < 10`。

# §7 · 逻辑运算符与优先级（P11）

## 课件原文（P11）

**Logical Operators**

- Operate on two or more logical expressions.
- Key Operators:
  - `and`: True if both expressions are true.
  - `or`: True if at least one expression is true.
  - `not`: Inverts the truth value of an expression.

**Truth Table Examples**

- **AND Operator:** `P and Q` is True if both P and Q are True.
- **OR Operator:** `P or Q` is True if at least one of P or Q is True.

**Order of Operations**

- Arithmetic operations are executed first, followed by comparison operations, and finally logical operations.
- **Tip:** Use parentheses to clarify order, e.g., `(1 + 3) > (2 + 5)` is more readable.

**Python Quirks**

- Python allows expressions like `(3 > 2) + (5 > 4)`, which resolves to `2`.
- Any non-zero number is considered True in logical operations.

## 人话版

**三个逻辑运算符**

| 运算符 | 中文 | 什么时候为真 |
|---|---|---|
| `and` | 与 | **两个都为真**才为真 |
| `or` | 或 | **至少一个为真**就为真 |
| `not` | 非 | 把真变假、假变真 |

**真值表（背下来，一辈子都用）**

**`P and Q`**

| P | Q | 结果 |
|---|---|---|
| True | True | **True** |
| True | False | False |
| False | True | False |
| False | False | False |

**`P or Q`**

| P | Q | 结果 |
|---|---|---|
| True | True | **True** |
| True | False | **True** |
| False | True | **True** |
| False | False | False |

**`not P`**

| P | 结果 |
|---|---|
| True | False |
| False | True |

> 💡 **记忆口诀**：`and` 很严格（一个不达标就否），`or` 很宽松（有一个达标就过）。
> 英文里 "you need ID **and** a ticket" 和 "pay by cash **or** card" 是同一个逻辑。

**优先级：算术 > 比较 > 逻辑**

课件那句话翻译成一张表（从高到低）：

```
1. 算术运算     + - * / // % **
2. 比较运算     < <= > >= == !=
3. 逻辑运算     not  →  and  →  or
```

课件推荐的写法 `(1 + 3) > (2 + 5)` 说明了这条规则：
**先算两边的加法，再比较**。虽然不加括号 Python 也算得对，但加了括号人一眼就能读懂。

## §7.2 🧪 短路求值：Python 什么时候「偷懒」

这是 `and` / `or` 最重要、也最容易被忽略的行为：**它们不一定把两边都算完**。

- `A and B`：**A 是假，就根本不去算 B**（反正结果是假）
- `A or B`：**A 是真，就根本不去算 B**（反正结果是真）

````python
def noisy(name, value):
    """一个有副作用的函数：被调用时会打印一行，然后返回 value"""
    print(f'   >> 求值了 {name}')
    return value

print('情形 1：False and ... —— 右边会被跳过吗？')
result = noisy('A', False) and noisy('B', True)
print('   结果 =', result)

print()
print('情形 2：True or ... —— 右边会被跳过吗？')
result = noisy('A', True) or noisy('B', False)
print('   结果 =', result)

print()
print('情形 3：两边都躲不掉的时候')
result = noisy('A', True) and noisy('B', True)
print('   结果 =', result)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-5 | `def noisy(name, value):` | 定义一个「会喊叫」的函数：每次被调用就打印一行。**这样我们就能看见谁被真的算过了** |
| 7 | `noisy('A', False) and noisy('B', True)` | 左边返回 False → `and` 立刻放弃，**右边根本不执行** |
| 12 | `noisy('A', True) or noisy('B', False)` | 左边返回 True → `or` 立刻放弃 |

**对照输出，你会看到：**

| 情形 | 屏幕上出现了几行「求值了」 |
|---|---|
| `False and ...` | **只有 1 行**（A）—— B 被跳过了 |
| `True or ...` | **只有 1 行**（A）—— B 被跳过了 |
| `True and ...` | **2 行**（A 和 B）—— 躲不掉 |

> ⚠️ **这个「偷懒」有两个后果，一好一坏。**
>
> **好处（要善用）**：可以防止崩溃。比如 `if len(a) > 0 and a[0] > 0:` ——
> 如果 a 是空的，右边根本不会执行，也就不会 `IndexError`。**把便宜的、能挡风险的判断写在左边。**
>
> **坏处（要警惕）**：如果右边有副作用（改文件、改全局变量、计数器 +1），
> **它可能没被执行**，而你的代码以为它执行了。
>
> 💡 一句话：**`and`/`or` 的右边是「有条件的执行」，不是「一定会执行」。**

## §7.3 🧪 优先级实测 + Python 的两个怪癖

课件的 Tip 和 Quirks 两段都可以直接实测。

````python
print('--- 优先级：算术 > 比较 > 逻辑 ---')
print('(1 + 3) > (2 + 5)       ->', (1 + 3) > (2 + 5))     # 课件的 Tip 例子
print('1 + 3 > 2 + 5           ->', 1 + 3 > 2 + 5)         # 不加括号，结果一样
print('1 + 3 > 2 + 5 and 1 < 2 ->', 1 + 3 > 2 + 5 and 1 < 2)

print()
print('--- 怪癖 1：(True) + (True) 真的是 2 ---')
print('(3 > 2) + (5 > 4)       ->', (3 > 2) + (5 > 4))     # 课件的 Quirk 例子

print()
print('--- 怪癖 2：and / or 返回的是「操作数本身」，不是 True/False ---')
print("0 or 'default'          ->", repr(0 or 'default'))
print("'abc' and 42            ->", repr('abc' and 42))
print("'' or [] or {}           ->", repr('' or [] or {}))
print("'abc' or 'never'        ->", repr('abc' or 'never'))

print()
print('--- 怪癖 3：任何非零数都算 True ---')
if 42:
    print('if 42:   → 走进了 if 分支')
if -3:
    print('if -3:   → 也走进了 if 分支（负数也是真！）')
if 0:
    print('这行永远不会打印')
if 0.0:
    print('这行也不会')
print('空字符串、空列表、空字典，全都是「假」:', bool(''), bool([]), bool({}))
print('但非空的都是「真」              :', bool('0'), bool([0]), bool({0: 0}))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4 | `1 + 3 > 2 + 5` | 算术先做 → `4 > 7` → `False`。**和加括号的结果一模一样**，证明优先级规则 |
| 5 | `... and 1 < 2` | 逻辑运算最后做。`False and True` → `False` |
| 9 | `(3 > 2) + (5 > 4)` | 课件的 Quirk。两个都是 True → `1 + 1` → **2** |
| 13-16 | `repr(...)` | `repr` 会**带上引号**打印，这样才能看清 `'default'` 是字符串 |
| 13 | `0 or 'default'` | ⚠️ `0` 是假 → `or` 返回**右边那个操作数本身**，也就是 `'default'`。**这不是 `True`！** |
| 14 | `'abc' and 42` | `'abc'` 是真 → `and` 返回右边 → `42`（不是 `True`） |
| 15 | `'' or [] or {}` | 全是假 → 返回**最后一个** → `{}` |
| 21-27 | `if 42:` / `if -3:` | 课件那句「Any non-zero number is considered True」的实测。**注意 `-3` 也是真** —— 判断真假只看「是不是 0」，不看正负 |
| 28-29 | `bool(...)` | 强制转换看真假。**空容器一律为假，非空一律为真** |

> ⚠️ **怪癖 2 是 Python 里最实用也最容易误读的特性。**
>
> `x = value or default` 是最常见的 Python 惯用法，意思是「value 如果为空就取 default」。
> 但你必须知道：**它返回的是 `value` 或 `default` 本身**。
> 所以 `count = 0` 时写 `count = count or 10` 会得到 **10**，而不是 0 —— 因为 0 是假！
>
> 📌 **勘误补充**：课件说 `(3 > 2) + (5 > 4)` 「resolves to 2」是**对的**，
> 但只说了结果、没说原因。原因是 `True` 在算术里就是 1 ——
> 这也是**唯一**一种你会看到布尔值和数字混算的场合，正常代码里不推荐这么写。

## 这一章你到底学会了什么

## 两个坑（本章实测过）

1. **浮点数不能用 `==` 比较。** `0.1 + 0.2 != 0.3`，`3*0.1 != 0.3`。要用 `math.isclose()`。
2. **`and` / `or` 右边不一定会执行**（短路），而且它们返回的是**操作数本身**而不是 `True`/`False`。

## 两个好习惯

1. **不确定优先级就加括号。** 括号不花钱，只赚可读性。
2. **写 `if` 之前先想清楚「这个条件什么时候是真」。** 如果说不清楚，就先在 notebook 里把它的真值表打出来（像 §7.1 那样）。

> 字符串、列表、元组、集合、字典、numpy 数组。**这些「容器」才是真正干活的工具。**
