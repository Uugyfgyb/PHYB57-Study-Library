# §1 · 函数的三件套（P1）

## 课件原文（P1）

**Key Concepts**

- **Function Definition:** A function in Python is a block of reusable code that performs a specific task. It is defined using the `def` keyword, followed by the function name, parameters in parentheses, and a colon.
- **Components of a Function:**
  1. **Function Header:** Starts with `def`, includes the function name and parameters.
  2. **Function Body:** Contains the statements that execute the function's task, indented for structure.
  3. **Return Statement:** Optional; returns the output of the function.

```python
def my_adder(a, b, c):
    """Function to sum three numbers"""
    return a + b + c

result = my_adder(1, 2, 3)   # Returns 6
```

**Best Practices**

- Use descriptive names for functions and variables.
- Comment extensively to clarify the function's purpose and logic.
- Regularly save your work.

## 人话版

**函数 = 「一台机器」**：喂进去几个东西（参数），吐出来一个东西（返回值）。

**三件套，一个都不能少：**

```python
def my_adder(a, b, c):        # ← ① 函数头：def + 名字 + 参数 + 冒号
    """三个数求和"""           # ← ② 函数体（缩进）：这里写要干的活
    return a + b + c          # ← ③ return：把结果交出去
```

| 部件 | 关键点 |
|---|---|
| ① **函数头** | `def` 是 define 的缩写；**结尾那个冒号别忘**；参数写在括号里，用逗号隔开 |
| ② **函数体** | **必须缩进**（4 个空格）。缩进就是 Python 的「大括号」 |
| ③ **`return`** | **可选的**。没有 `return` 的函数返回 `None` |

> ⚠️ **定义 ≠ 调用。**
>
> `def my_adder(a, b, c):` 这一行**只是「造机器」**，不会执行任何计算。
> 真正让它干活的是 `my_adder(1, 2, 3)` —— **写了括号才叫调用。**
>
> 这是新手最常见的困惑：「我明明定义了，为什么什么都没算？」

**那个三引号字符串是什么？**

`"""Function to sum three numbers"""` 叫 **docstring（文档字符串）**。
它不参与计算，但会被保存进函数里 —— C1 里我们用 `math.factorial.__doc__` 看过这个机制。
**在 Jupyter 里敲 `my_adder?` 就能看到它。**

> 💡 课件把这句 docstring 写进了例子，说明它**不是装饰，是规范**。
> 写函数就顺手写一句它干什么，三个月后的你会感谢现在的你。

## §1.1 课件的例子：定义 + 调用

````python
def my_adder(a, b, c):
    """Function to sum three numbers"""
    return a + b + c


result = my_adder(1, 2, 3)   # Returns 6
print('result =', result)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `def my_adder(a, b, c):` | **函数头**。`a b c` 叫**形参**（形式参数），现在还没有值，只是三个「占位名字」 |
| 2 | `"""Function to sum three numbers"""` | docstring。**函数的第一句话如果是字符串，它就自动成为文档** |
| 3 | `return a + b + c` | 把三个参数加起来，**交出去**，同时**结束函数** |
| 6 | `my_adder(1, 2, 3)` | **调用**：把 1、2、3 按位置传给 `a`、`b`、`c`（这叫**位置参数**），执行，拿到返回值 6 |
| 7 | `print('result =', result)` | `result` 就是那个 6 |

> 💡 **形参 vs 实参**（两个词经常被混用）：
>
> | 名字 | 在哪 | 例子 |
> |---|---|---|
> | **形参** parameter | 定义时的 `a, b, c` | 只是名字，还没有值 |
> | **实参** argument | 调用时的 `1, 2, 3` | 真正的值 |
>
> 位置是**一一对应**的：第 1 个实参给第 1 个形参，以此类推。**顺序搞反了不会报错，只会算错。**

## §1.2 🧪 函数到底是什么：五个小实验

课件只给了一行调用。我们把函数的「脾气」一次性摸清楚。

````python
print('--- 1. 函数也是对象：它有自己的类型和名字 ---')
print('  type(my_adder)  =', type(my_adder))
print('  my_adder.__name__ =', repr(my_adder.__name__))
print('  docstring       =', repr(my_adder.__doc__))
print()

print('--- 2. 括号是「调用」，没有括号是「这个东西本身」 ---')
print('  my_adder         ->', my_adder, '  <- 函数的地址，不是结果')
print('  my_adder(1,2,3)  ->', my_adder(1, 2, 3), '  <- 这才调用')
print()

print('--- 3. 同一个函数可以反复调用，参数随便换 ---')
for args in [(1, 2, 3), (10, 20, 30), (-1, 0, 1), (0.5, 0.25, 0.125)]:
    print(f'  my_adder{args} = {my_adder(*args)}')
print('  （*args 的意思是「把元组拆开当参数传进去」，C2 §4 的解包）')
print()

print('--- 4. 参数个数不对会怎样？ ---')
try:
    my_adder(1, 2)
except TypeError as e:
    print('  少给一个 -> TypeError:', e)
try:
    my_adder(1, 2, 3, 4)
except TypeError as e:
    print('  多给一个 -> TypeError:', e)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2-4 | `type(my_adder)` | → `<class 'function'>`。**函数是一个真正的对象**，不是语法糖 |
| 3 | `my_adder.__name__` | → `'my_adder'`。函数记得自己的名字 |
| 4 | `my_adder.__doc__` | → `'Function to sum three numbers'`。**docstring 真的被存下来了** |
| 8 | `print(my_adder)` | ⚠️ **没有括号时，打印的是「函数对象本身」**，形如 `<function my_adder at 0x...>` |
| 9 | `my_adder(1,2,3)` | 有括号才是调用 |
| 13-14 | `my_adder(*args)` | `*args` 把元组 `(1,2,3)` **拆成三个独立的参数**。这是 C2 §4 解包的直接应用 |
| 18-23 | 参数个数不对 | 多一个少一个都会报 `TypeError`，**报错信息里会明确告诉你「要 3 个，给了 2 个」** |

> ⚠️ **第 8 行是新手最容易犯的错**：写 `print(my_adder)` 想看到结果，
> 结果打出来一串 `<function ... at 0x7f...>`。
> **看到 `<function ...>` 就说明你忘了加括号。**
>
> 💡 **`type(my_adder)` 这一行很重要** —— 它证明了「函数就是一种普通的值」。
> 正因为它是一种值，才可以**被赋值给变量、被当参数传、被当返回值返回**（§8 的全部内容）。

# §2 · 输入参数与输出参数（P2）

## 课件原文（P2）

**Input Arguments**

- Functions can accept multiple input arguments, which are variables provided by the user or another function.
  ```python
  def multiply(x, y):
      return x * y
  result = multiply(4, 5)   # Returns 20
  ```

**Output Parameters**

- Functions may return multiple values as a **tuple**.
  ```python
  def my_trig_sum(a, b):
      out1 = np.sin(a) + np.cos(b)
      out2 = np.sin(b) + np.cos(a)
      return out1, out2
  c, d = my_trig_sum(2, 3)
  ```

**Error Handling**

- Python functions assume input arguments will be of the expected type. If not, errors like `TypeError` may occur.
  ```python
  def my_adder(a, b, c):
      return a + b + c
  result = my_adder(1, 2, "three")   # Raises TypeError
  ```

## 人话版

**输入：可以多个**

`multiply(x, y)` 收两个，`my_adder(a, b, c)` 收三个。**要几个写几个，用逗号隔开。**

**输出：可以多个（靠元组打包）**

```python
return out1, out2      # ← 这不是「返回两个值」
                       #    而是「把两个值打包成一个元组，再返回」
```

所以 `c, d = my_trig_sum(2, 3)` 能工作，靠的正是 **C2 §4 学的元组解包**。

> 💡 **一句话**：Python 的函数**永远只返回一个东西**。
> 想返回多个，就把它们**打包成元组**。语法上看起来像返回多个，其实不是。

**错误处理：Python 不管类型，出事才报错**

课件那句话的意思是：**你在函数里写的每一行，Python 都假定参数类型是对的。**
它不会在进门时检查「你给的是不是数字」，而是等真正算到那一行才炸。

```python
my_adder(1, 2, "three")
# a + b = 3        ✓ 没问题
# 3 + "three"      ✗ 到这一步才炸：TypeError
```

## §2.1 输入参数：课件两个例子

````python
import numpy as np     # 下面的 my_trig_sum 要用


def multiply(x, y):
    return x * y

result = multiply(4, 5)   # Returns 20
print('multiply(4, 5) =', result)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `import numpy as np` | 导入 numpy。**注意这行写在 notebook 顶层，不在函数里** —— 函数内部可以直接用外面导入的模块 |
| 4 | `def multiply(x, y):` | 两个参数 |
| 5 | `return x * y` | 乘起来交出去 |
| 7 | `multiply(4, 5)` | → `20`（和课件一致） |

> 💡 **函数体里能不能用外面定义的变量？** 能。
> 但如果那是**全局变量**，就有 §5 要讲的那堆麻烦。**记住一条**：
> 函数需要的所有东西，**最好都从参数进来**，不要在函数里偷偷读外面的变量。

## §2.2 输出参数：多个返回值其实是元组

````python
def my_trig_sum(a, b):
    out1 = np.sin(a) + np.cos(b)
    out2 = np.sin(b) + np.cos(a)
    return out1, out2


c, d = my_trig_sum(2, 3)
print('c =', c)
print('d =', d)
print()

print('--- 证据：不加解包，看到的就是一个元组 ---')
raw = my_trig_sum(2, 3)
print('  my_trig_sum(2, 3) =', raw)
print('  type(...)         =', type(raw).__name__)
print('  len(...)          =', len(raw))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-4 | `def my_trig_sum(a, b):` | 函数头 + 两个中间变量 `out1` `out2` |
| 2 | `out1 = np.sin(a) + np.cos(b)` | ⚠️ **`a` 和 `b` 是弧度**。`np.sin(2)` 是 sin 2 弧度 ≈ 0.909 |
| 4 | `return out1, out2` | ⚠️ **这一行返回的不是两个值，是一个二元组** `(out1, out2)` |
| 7 | `c, d = my_trig_sum(2, 3)` | **解包**：把元组拆给 `c` 和 `d` |
| 12-14 | `raw = my_trig_sum(2, 3)` | 故意不解包，看它到底是什么 → `type` 是 `tuple`，`len` 是 2 |

**跑出来的数字长这样：**

| 表达式 | 值 | 怎么来的 |
|---|---|---|
| `np.sin(a)` = `np.sin(2)` | 0.9092974268256817 | sin 2 **弧度** |
| `np.cos(b)` = `np.cos(3)` | −0.9899924966004454 | cos 3 弧度 |
| **`out1` = `c`** | **−0.0806950697747637** | 0.9092974268256817 − 0.9899924966004454 |
| `np.sin(b)` = `np.sin(3)` | 0.1411200080598672 | sin 3 弧度 |
| `np.cos(a)` = `np.cos(2)` | −0.4161468365471424 | cos 2 弧度 |
| **`out2` = `d`** | **−0.2750268284872752** | 0.1411200080598672 − 0.4161468365471424 |

> 💡 **这张表就是「函数调试」的标准动作**：把函数体里每一步拆开单独算一遍，
> 和你以为的对齐。**不要靠心算，也不要靠猜。**
>
> ⚠️ 一个容易看花眼的地方：`out1` 用的是 **sin(a) + cos(b)**，`out2` 用的是 **sin(b) + cos(a)** ——
> **两个三角函数的自变量是交叉的**。这种「故意交叉」的写法在物理里很常见（比如对称化处理），
> 但也是最容易抄错的地方。
>
> 📎 **numpy 2.x 的一个显示变化**：下面第 13 行 `print('  my_trig_sum(2, 3) =', raw)` 的输出会是
> `(np.float64(-0.0806950697747637), np.float64(-0.2750268284872752))` ——
> 外面那层 `np.float64(...)` 是 numpy 2.0 之后**打印元组时的表示法**，
> 说明这两个数**不是 Python 原生 float，而是 numpy 的标量**。
> 它们能和普通数字一样运算，只是 `type()` 查出来是 `numpy.float64`。

## §2.3 把四个数值单独验一遍

上面那张表里的数字是**算出来的**，不是抄的。这一格把中间量全部打印出来，你可以逐项对照。

````python
a, b = 2, 3
print(f'  a = {a}, b = {b}')
print(f'  np.sin(a) = sin({a}) = {np.sin(a):.6f}')
print(f'  np.cos(a) = cos({a}) = {np.cos(a):.6f}')
print(f'  np.sin(b) = sin({b}) = {np.sin(b):.6f}')
print(f'  np.cos(b) = cos({b}) = {np.cos(b):.6f}')
print()
print(f'  out1 = sin(a) + cos(b) = {np.sin(a):.6f} + ({np.cos(b):.6f}) = {np.sin(a) + np.cos(b):.6f}')
print(f'  out2 = sin(b) + cos(a) = {np.sin(b):.6f} + ({np.cos(a):.6f}) = {np.sin(b) + np.cos(a):.6f}')
print()
print('  和 my_trig_sum(2, 3) 的输出对一下:', my_trig_sum(2, 3))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `a, b = 2, 3` | 打包/解包（C2 §4），一行定义两个变量 |
| 2-5 | 四个值单独打印 | 每次都用 `:.6f` 保留 6 位小数，**把心算和实测对齐** |
| 8-9 | 两项分别相加 | 明确写成 `sin(a)+cos(b)` 和 `sin(b)+cos(a)`，避免「记错公式」 |
| 11 | 和函数输出对照 | 用同一组数再调一次函数，**两边必须一致** |

**这张表就是「调试函数」的标准动作**：把函数的每一步拆开单独算，和你以为的对齐。

> 💡 **这里真正想教的是**：看到函数的输出和你预期不符时，
> **第一反应应该是「把中间量打出来」，而不是「Python 算错了」。**
> 90% 的情况是你自己记错了公式或者搞反了参数。
>
> ⚠️ 再强调一次：`np.sin` 的参数是**弧度**。`sin(2)` 是 sin 2 rad ≈ 0.909，
> **不是** sin 2° ≈ 0.0349。想做角度就得先 `np.radians(2)`。

## §2.4 类型错误：课件 P2 的最后一个例子

````python
def my_adder(a, b, c):
    return a + b + c


print('--- 正常情况 ---')
print('  my_adder(1, 2, 3) =', my_adder(1, 2, 3))

print()
print('--- 课件 P2 的错误例子：把字符串混进来 ---')
try:
    result = my_adder(1, 2, "three")
except TypeError as e:
    print('  TypeError:', e)

print()
print('--- 为什么会这样？逐步看 ---')
a, b, c = 1, 2, "three"
print('  a + b     =', a + b, '   <- 两个数字，没问题')
try:
    print('  (a+b) + c =', (a + b) + c)
except TypeError as e:
    print('  (a+b) + c -> TypeError:', e, '  <- 到这里才炸')
print()
print('  int + str 本来就是非法的：')
try:
    3 + "three"
except TypeError as e:
    print('    3 + "three" -> TypeError:', e)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 10-11 | `my_adder(1, 2, "three")` | 课件 P2 的例子。报 `TypeError`，**而且信息里会说 `unsupported operand type(s) for +: 'int' and 'str'`** |
| 16-21 | 拆开算 | 证明 `a + b` 是好的，**错误发生在 `+ c` 那一步** —— 也就是课件说的「Python 不提前检查类型，算到才炸」 |
| 24-25 | `3 + "three"` | 直接演示：**两种类型不能用 `+` 相加**（字符串只能加字符串） |

> 💡 **`+` 对不同类型的含义完全不同：**
>
> | 两边都是 | `+` 的含义 |
> |---|---|
> | 数字 + 数字 | 算术加法 |
> | 字符串 + 字符串 | 拼接 |
> | 列表 + 列表 | 合并（C2 §3） |
> | **数字 + 字符串** | ❌ **TypeError** |
>
> **这是 Python 和 JavaScript 的一个重要区别**（JS 会偷偷把 `1 + "2"` 变成 `"12"`）。
> Python 选择「宁可报错，也不猜你的意思」—— 这是好事。

# §3 · 多返回值、默认参数、函数组合（P3）

## 课件原文（P3）

**Multiple Outputs**

- Functions can return multiple outputs, typically as a tuple, allowing the user to unpack the values into separate variables.
  ```python
  def calculate(a, b):
      return a + b, a * b
  sum_result, product_result = calculate(5, 10)
  ```

**Default Parameters**

- Functions can have default parameter values, which are used if no argument is provided during the function call.
  ```python
  def greet(name="World"):
      print(f"Hello, {name}!")
  greet()          # Prints "Hello, World!"
  greet("Alice")   # Prints "Hello, Alice!"
  ```

**Composition of Functions**

- Functions can be composed, meaning the output of one function can be used as the input for another.
  ```python
  def add(a, b):
      return a + b
  def square(x):
      return x * x
  result = square(add(2, 3))   # Returns 25
  ```

## 人话版

三件事，一个一个说：

**① 多返回值 = 打包成元组**（§2 已经讲过原理）

```python
return a + b, a * b     # 实际返回的是元组 (a+b, a*b)
```

**② 默认参数：调用时可以不传**

```python
def greet(name="World"):   # ← 这里给了个默认值
    ...
greet()          # 不传 → name 用默认值 "World"
greet("Alice")   # 传了 → name 用 "Alice"
```

⚠️ **默认参数必须写在普通参数的后面。** `def f(a=1, b)` 是**语法错误**。

**③ 函数组合：嵌套调用**

```python
square(add(2, 3))
```

**执行顺序是从里往外**：先算 `add(2,3)` = 5，再把 5 喂给 `square` → 25。

这和数学里的 f(g(x)) 一模一样。

> 💡 **为什么函数组合重要？** 因为它让你可以**用积木搭积木**：
> `add` 和 `square` 各自都很简单，但组合起来能表达复杂逻辑，
> **而且每一个都还能单独测试。** 这是「可维护代码」的核心思想。

## §3.1 三个例子一起跑

````python
print('=== ① 多返回值 ===')


def calculate(a, b):
    return a + b, a * b


sum_result, product_result = calculate(5, 10)
print('  sum_result     =', sum_result)
print('  product_result =', product_result)
print('  不解包看看      =', calculate(5, 10))

print()
print('=== ② 默认参数 ===')


def greet(name="World"):
    print(f"  Hello, {name}!")


greet()            # Prints "Hello, World!"
greet("Alice")     # Prints "Hello, Alice!"

print()
print('=== ③ 函数组合 ===')


def add(a, b):
    return a + b


def square(x):
    return x * x


result = square(add(2, 3))   # Returns 25
print('  square(add(2, 3)) =', result)
print('  拆开验证：add(2,3) =', add(2, 3), '，square(5) =', square(5))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-5 | `return a + b, a * b` | 返回一个二元组。**注意 `return` 后面没有括号**（加了也合法，`return (a+b, a*b)` 一模一样） |
| 8 | `sum_result, product_result = calculate(5, 10)` | 解包 → 15 和 50 |
| 9 | `calculate(5, 10)` | 不解包，看到真面目 `(15, 50)` |
| 15 | `def greet(name="World"):` | **默认参数**写在形参后面，用 `=` 给值 |
| 19 | `greet()` | 不传参 → 用默认值 |
| 20 | `greet("Alice")` | 传了就用传的 |
| 32 | `square(add(2, 3))` | **从里往外**：先 `add(2,3)`=5，再 `square(5)`=25（和课件一致） |
| 33 | 拆开验证 | 把中间结果打出来，确认「从里往外」的顺序 |

> ⚠️ **默认参数的一个大坑：默认值只在「定义函数时」算一次。**
>
> 如果默认值是**可变对象**（列表、字典），所有调用会**共享**同一个：
>
> ```python
> def bad(x, lst=[]):     # ⚠️ 危险！
>     lst.append(x)
>     return lst
>
> bad(1)   # → [1]
> bad(2)   # → [1, 2]   ← 你以为会得到 [2]！
> ```
>
> **正确写法**是默认值用 `None`，进函数后再建新的：
> ```python
> def good(x, lst=None):
>     if lst is None:
>         lst = []
>     lst.append(x)
>     return lst
> ```
> 下面这一格就实测这个坑。

## §3.2 ⚠️ 默认参数的可变对象陷阱

````python
print('--- 错误示范：默认值是空列表 ---')


def bad_append(x, lst=[]):
    lst.append(x)
    return lst


print('  bad_append(1) =', bad_append(1))
print('  bad_append(2) =', bad_append(2))
print('  bad_append(3) =', bad_append(3))
print()

print('--- 正确示范：默认值用 None ---')


def good_append(x, lst=None):
    if lst is None:
        lst = []          # 每次调用都新建一个
    lst.append(x)
    return lst


print('  good_append(1) =', good_append(1))
print('  good_append(2) =', good_append(2))
print('  good_append(3) =', good_append(3))
print()
print('  显式传入列表时，仍然会就地修改传进来的那个：')
mine = [100]
print('    good_append(200, mine) =', good_append(200, mine))
print('    mine 现在是            =', mine, '  <- 被改了（这是预期行为）')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-6 | `def bad_append(x, lst=[])` | 默认值是一个**列表对象**，它在 `def` 执行的那一刻被创建，**之后所有调用共用同一个** |
| 9-11 | 连续三次调用 | ⚠️ 结果是 `[1]`、`[1,2]`、`[1,2,3]` —— **列表在累积**，完全不是预期 |
| 17-21 | `lst=None` 的写法 | `None` 是不可变的，每次进函数都新建一个空列表 → 每次都是独立的 |
| 25-27 | 显式传列表 | 这时函数**就地修改你传进去的那个列表** —— 这是**故意的**行为，不是 bug |

> ⚠️ **这个坑在真实项目里的表现非常阴险**：单测每次单独跑都通过，
> 一旦连续跑多次就出错。**看到默认值是 `[]` 或 `{}`，立刻警惕。**
>
> 💡 **记住这条规则**：
> **默认值只用「不可变」的东西** —— 数字、字符串、元组、`None`。
> 需要可变容器，就用 `None` + 在函数体里新建。

## §3.3 📌 勘误：课件「说一套做一套」—— 要求写 docstring，自己的例子却全都没写

课件 P1 的例子里，docstring 是这样写的：

```python
def my_adder(a, b, c):
    """Function to sum three numbers"""
    return a + b + c
```

**这个位置本身是对的**（docstring 就该紧跟在 `def` 那一行的下面）。

**问题出在后面**：从 P2 开始，课件给的每一个函数例子**都不带 docstring 了** ——
而课件 P1 自己的 Best Practices 里明明写着 "Comment extensively"。

**这不是错，是「说了但没做」。** 真正的问题在于：**很多人以为 docstring 只是个注释，随便写。**

它不是。下面实测。

````python
def with_doc(a, b):
    """把两个数相加。

    Parameters
    ----------
    a, b : float
        要相加的两个数。

    Returns
    -------
    float
        a + b。
    """
    return a + b


def without_doc(a, b):
    return a + b


print('--- 有 docstring 的函数 ---')
print(with_doc.__doc__)
print()
print('--- 没有 docstring 的函数 ---')
print(repr(without_doc.__doc__), '  <- 是 None！')
print()
print('--- 在 Jupyter 里，这两种的差别 ---')
print('  敲 with_doc?     -> 弹出完整说明（上面那段）')
print('  敲 without_doc?  -> 只显示一行签名，什么说明都没有')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-13 | `def with_doc(a, b):` | 多行 docstring 的标准格式：一句话总结 + 参数说明 + 返回值说明 |
| 16-17 | `def without_doc(a, b):` | 没有 docstring |
| 21 | `with_doc.__doc__` | 打印出完整的说明文本 |
| 25 | `without_doc.__doc__` | ⚠️ 是 **`None`**，不是空字符串 |
| 29-30 | `?` 的效果 | 有 docstring 的函数敲 `?` 会弹出完整文档；没有的只有一行签名 |

> 📌 **勘误说明**：课件 P1 把 docstring 列进了「函数的三件套」之一（这个定位是对的），
> 但从 P2 开始的所有例子都**把 docstring 省掉了**，而 Best Practices 又要求 "Comment extensively"。
> **说了却没做，是教学材料里最常见的自相矛盾。**
>
> 💡 **实际要求**：作业和小项目里，
> - **每个函数**至少写**一行** docstring：它干什么
> - 参数多或者返回值复杂时，再补参数说明
>
> 写 docstring 的直接好处：**在 Jupyter 里敲 `?` 就能看到**，不用翻回定义处。

# §4 · 局部变量（P4）

## 课件原文（P4）

- **Local Variables** are variables that are declared **inside a function**.
- These variables are accessible **only within that function** and have their **own memory block** separate from the rest of the program.
- The value of a local variable is **not affected by and does not affect** variables outside the function, **even if they share the same name**.

```python
def my_adder(a, b, c):
    out = a + b + c
    print(f'The value of out within the function is {out}')
    return out

out = 1
d = my_adder(1, 2, 3)
print(f'The value of out outside the function is {out}')
```

**Output**

- Inside the function: `out = 6`
- Outside the function: `out = 1`

**Key Points**

- When the function is executed, a new memory block is created.
- Variables inside the function do not interfere with variables outside it.

## 人话版

**函数有自己的一间「屋子」。**

```
外面（全局）              函数里面（局部）
─────────────            ─────────────
out = 1                  out = a + b + c   <- 同名，但完全是两个变量
                         （函数一结束，这个 out 就消失了）
```

**三条规则：**

| 规则 | 意思 |
|---|---|
| 函数里定义的变量**只活在函数里** | 出了函数就没人认识它 |
| 函数**同名**也不冲突 | 里面一个 `out`，外面一个 `out`，互不干扰 |
| 函数**能读**外面的变量 | 但「读」和「写」规则不一样（§5 的重点） |

> 💡 课件说「a new memory block is created（会新建一个内存块）」，
> 这个说法对理解很有帮助：**每次调用函数，都开一间新屋子；函数返回，屋子就拆了。**
>
> 所以函数里定义的变量**不会在两次调用之间保留**（想保留就得用默认参数或者对象属性）。

## §4.1 课件的例子：同名但互不干扰

````python
def my_adder(a, b, c):
    out = a + b + c
    print(f'The value of out within the function is {out}')
    return out


out = 1
d = my_adder(1, 2, 3)
print(f'The value of out outside the function is {out}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-4 | `def my_adder(a, b, c):` | 函数体里有个 `out` |
| 2 | `out = a + b + c` | ⚠️ **这个 `out` 是「屋里」的**，函数一结束就消失 |
| 6 | `out = 1` | ⚠️ **这个 `out` 是「屋外」的**，和上面那个毫无关系 |
| 7 | `d = my_adder(1, 2, 3)` | 进函数：里面的 `out` = 6，打印 6；然后函数结束，**屋里的 `out` 被销毁** |
| 8 | `print(out)` | 外面的 `out` **从始至终都是 1** |

**输出与课件完全一致**：

| 位置 | 值 |
|---|---|
| 函数内 | `out = 6` |
| 函数外 | `out = 1` |

> 💡 **注意第 7 行的 `d`**：函数返回的 6 被存进了 `d`。
> **这是函数「把结果带出屋子」的唯一正规通道** ——
> 屋里的东西出不来，除非你 `return` 它。

## §4.2 🧪 证明「函数里的变量真的消失了」

````python
def demo():
    inside_only = 'I am local'
    print('  函数内能看见 inside_only:', repr(inside_only))
    return None


demo()
print()
try:
    print(inside_only)
except NameError as e:
    print('  函数外看不见 ->', type(e).__name__ + ':', e)
print()

print('--- 函数里的参数也一样是局部的 ---')


def show_local(a):
    print('  函数内 a =', a)
    del a          # 连参数都能删
    print('  删掉之后，函数内还有 a 吗？', 'a' in dir())


show_local(999)

try:
    print(a)
except NameError as e:
    print('  函数外看不见 a ->', type(e).__name__ + ':', e)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2 | `inside_only = 'I am local'` | 在屋内造一个变量 |
| 3 | `print(inside_only)` | 屋内能看见 ✓ |
| 9-12 | `print(inside_only)` 在屋外 | ❌ `NameError: name 'inside_only' is not defined` |
| 18-21 | `def show_local(a):` | **函数的参数 `a` 也是局部的** —— 它和 `inside_only` 的地位完全一样 |
| 20 | `del a` | 连参数都能删掉。**这证明参数只是「预先生好的局部变量」** |
| 25-28 | 屋外访问 `a` | 同样 `NameError` |

> 💡 **一个重要的心智模型**：
> `def f(a, b):` 里那个 `a`、`b`，**等价于**在函数第一行写了 `a = <传进来的值>`。
> 它们不是特殊的「参数」，就是**预先赋值好的局部变量**。
>
> ⚠️ 所以「`del a` 之后 `a` 就没了」这件事是成立的，虽然实际代码里没人这么写。

# §5 · 全局变量（P5–P6）

## 课件原文（P5）

- **Global Variables** are declared **outside** of functions and are accessible anywhere in the program.
- They can be **read and modified** both inside and outside functions, **using the `global` keyword**.

**Common Issues**

- Without declaring a variable as global, attempts to modify it inside a function can lead to errors like `UnboundLocalError`.

**Example Without Global Keyword**

```python
n = 42

def func():
    print(f'Within function: n is {n}')
    n = 3
    print(f'Within function: n is now {n}')

func()
```

**Error:** `UnboundLocalError: local variable 'n' referenced before assignment`

**Solution Using Global Keyword**

```python
n = 42

def func():
    global n
    print(f'Within function: n is {n}')
    n = 3
    print(f'Within function: n is now {n}')

func()
print(f'Outside function: n is {n}')
```

**Output**

- Inside function: `n = 42`, then `n = 3`
- Outside function: `n = 3`

## 人话版

**先看那个「诡异的报错」：为什么读 `n` 会报 `UnboundLocalError`？**

这段代码看起来完全正常：

```python
n = 42
def func():
    print(n)   # <- 这里明明只是想读一下外面的 n
    n = 3      # <- 这里赋值
```

**但它在 `print(n)` 那一行就炸了。** 原因是 Python 的一条硬规则：

> **只要函数体里出现了对 `n` 的赋值，`n` 就是「局部变量」—— 整个函数都是，包括赋值之前的那几行。**

所以执行到 `print(n)` 时，Python 认为「你要用局部的 `n`」，但局部 `n` 还没赋值 -> **`UnboundLocalError`**。

> ⚠️ **报错信息 `local variable 'n' referenced before assignment`
> 直译就是「局部变量 n 在被赋值之前就被引用了」。**
> 这句话已经把原因说得很清楚了 —— **关键是看懂「local（局部）」这个词**。

**解决办法：用 `global` 声明**

```python
def func():
    global n     # <- 告诉 Python：这个 n 说的是外面那个
    ...
```

加了这一行，函数里的 `n` 就真的指向全局的 `n` 了 —— **读和写都是。**

| 操作 | 不加 `global` | 加了 `global` |
|---|---|---|
| 只**读**全局变量 | 可以 | 可以 |
| **改**全局变量 | 报 `UnboundLocalError`（或改的是新的局部变量） | 改的就是全局那个 |

## §5.1 复现课件那个错误

````python
n = 42


def func():
    print(f'Within function: n is {n}')
    n = 3
    print(f'Within function: n is now {n}')


print('--- 照课件 P5 原样写 ---')
try:
    func()
except UnboundLocalError as e:
    print('  报错 -> UnboundLocalError:', e)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `n = 42` | **全局变量**：定义在所有函数之外 |
| 4-7 | `def func():` | 函数体里**既有读 `n`（第 5 行）又有写 `n`（第 6 行）** |
| 5 | `print(f'... {n}')` | ⚠️ **报错就发生在这里**，而不是第 6 行 |
| 6 | `n = 3` | 这一行「提前宣判」了 `n` 是局部的 —— **哪怕它写在第 5 行之后** |
| 12-13 | `try/except` | 接住错误并打印 |

**报错原文**：`UnboundLocalError: cannot access local variable 'n' where it is not associated with a value`

（Python 3.11 之后措辞略有变化，课件的 `local variable 'n' referenced before assignment` 是旧版措辞。**含义完全一样。**）

> ⚠️ **这是 Python 里最反直觉的规则之一，值得反复看几遍。**
>
> **判定顺序是「静态」的**：Python 在**执行函数之前**就扫一遍整个函数体，
> 只要看到 `n = ...`，就认定 `n` 是局部的。**跟那一行的位置无关。**
>
> 💡 **对比一下：其他语言（比如 C）是按位置判断的**，所以从别的语言转过来的人特别容易栽在这。
>
> 📌 顺便记一下报错措辞的版本差异：
>
> | Python 版本 | 报错文本 |
> |---|---|
> | 3.10 及更早 | `UnboundLocalError: local variable 'n' referenced before assignment` |
> | 3.11 及更新 | `UnboundLocalError: cannot access local variable 'n' where it is not associated with a value` |
>
> **背报错信息时要认「意思」，不要认「字符串」。**

## §5.2 用 `global` 解决

````python
n = 42


def func():
    global n
    print(f'Within function: n is {n}')
    n = 3
    print(f'Within function: n is now {n}')


func()
print(f'Outside function: n is {n}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 5 | `global n` | ⚠️ **关键的一行**。它的位置在函数体最前面（约定俗成），告诉 Python「下面说的 `n` 是全局那个」 |
| 6 | `print(...)` | 读到全局的 42 |
| 7 | `n = 3` | **改的是全局的 `n`** —— 不再新建局部变量 |
| 12 | `print(f'Outside function: n is {n}')` | 外面的 `n` 真的变成了 3 |

**输出与课件完全一致**：

| 位置 | 值 |
|---|---|
| 函数内第一次 | `n = 42` |
| 函数内第二次 | `n = 3` |
| 函数外 | `n = 3`（**被函数改掉了**） |

> ⚠️ **这就是 `global` 的「威力」，也是它的「危险」**：
> 一个函数**偷偷改了外面的状态**，而且从调用处完全看不出来。
> 读到 `func()` 这一行的人，不会想到 `n` 会被改。
>
> 💡 **更好的做法**（课件 P6 的 Best Practices 说的就是这件事）：
>
> ```python
> def func(n):        # 把 n 当参数传进来
>     ...
>     return 3        # 把新值返回出去
>
> n = func(n)         # 调用处一眼就能看到 n 被更新了
> ```
>
> **同样的功能，但「输入」和「输出」都写在函数签名上了。** 这叫**显式优于隐式**。

## §5.3 「能读不能改」是什么意思？一个更清楚的实验

课件只给了报错的那个例子。下面这个实验把规则划得更清楚。

````python
counter = 0
CONFIG = {'name': 'PHYB57', 'debug': False}
readonly_list = [1, 2, 3]

print('--- 只读全局变量：完全没问题 ---')


def read_only():
    print('  能读到 counter =', counter)
    print('  能读到 CONFIG  =', CONFIG)
    print('  能读到列表长度  =', len(readonly_list))


read_only()

print()
print('--- 想「重新赋值」全局的不可变变量：必须 global ---')


def try_change_bad():
    counter = counter + 1     # ⚠️ 这一行会让 counter 变成局部变量
    return counter


try:
    try_change_bad()
except UnboundLocalError as e:
    print('  报错 ->', e)


def change_good():
    global counter
    counter = counter + 1
    return counter


change_good()
change_good()
print('  用了 global 之后 counter =', counter)

print()
print('--- 但「就地修改」可变对象，根本不需要 global！ ---')


def mutate_dict():
    CONFIG['debug'] = True        # 没有 global，也没报错
    readonly_list.append(4)       # 同样可以


mutate_dict()
print('  CONFIG 现在是  =', CONFIG)
print('  readonly_list 现在是 =', readonly_list)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 9-12 | `read_only()` | **只读**全局变量：毫无问题 |
| 18-20 | `counter = counter + 1` | ⚠️ 这一行包含**赋值**，所以 `counter` 被判为局部 -> `UnboundLocalError` |
| 28-31 | `global counter` | 加了 `global` 之后就能改了 |
| 39-41 | `CONFIG['debug'] = True` | ⚠️ **这一行没有 `global`，却成功了** |

**为什么最后那个能成功？** 因为它**不是赋值给 `CONFIG`**，而是「**取到 `CONFIG` 指向的那个字典，改它的内容**」。

| 写法 | 是「赋值给变量」吗 | 需要 `global` 吗 |
|---|---|---|
| `counter = counter + 1` | 是 | **需要** |
| `CONFIG['debug'] = True` | 否（改的是字典里面） | **不需要** |
| `readonly_list.append(4)` | 否（调方法改内容） | **不需要** |
| `CONFIG = {}` | 是 | **需要** |

> ⚠️ **这条区别和 C2 §3 的「`b = a` 不复制」是同一个道理**：
> **变量的名字**和**变量指向的对象**是两回事。
> `global` 管的是「名字」，管不到「对象的内容」。
>
> 💡 **这条规则既是便利也是陷阱**：一个函数可以不加声明就悄悄改掉你的全局字典，
> 从调用处完全看不出来。**所以还是那句话 —— 能当参数传就别用全局。**

## §5.4 课件 P6：三条 Best Practices

课件 P6 的三条建议，逐条翻译：

| 课件原文 | 人话 | 怎么做 |
|---|---|---|
| **Minimize Global Variables** | 全局变量能少则少 | 用参数传入、用返回值传出 |
| **Use Functions Effectively** | 发挥局部变量的作用 | 中间结果都放函数里，别污染全局 |
| **Global Keyword Use** | `global` 要谨慎 | **只在「确实需要一个全局计数器/配置」时才用** |

**为什么课件说全局变量「Overuse can lead to complex dependencies and harder-to-debug code」？**

因为全局变量让**函数的行为依赖于「之前发生过什么」** ——
同一行调用，在程序的不同时刻会给出不同结果：

````python
state = 0


def step_a():
    global state
    state = state + 1
    return state


def step_b():
    global state
    state = state * 10
    return state


print('--- 症状 1：同一个调用，结果取决于「被调用过几次」 ---')
state = 0
print('   第 1 次 step_a() ->', step_a())
print('   第 2 次 step_a() ->', step_a())
print('   第 3 次 step_a() ->', step_a())
print('   <- 一模一样的调用，结果 1、2、3 全不一样！')
print()

print('--- 症状 2：同一个调用，结果取决于「前面是谁」 ---')
state = 0
print('   先 a 后 b: a ->', step_a(), '  b ->', step_b())
state = 0
print('   先 b 后 a: b ->', step_b(), '  a ->', step_a())
print('   <- 两次的 step_a() 都是 1，但 step_b() 是 10 和 0')
print()

print('--- 换成「纯函数」：只依赖参数 ---')


def step_a_pure(s):
    return s + 1


def step_b_pure(s):
    return s * 10


print('   第 1 次 step_a_pure(0) ->', step_a_pure(0))
print('   第 2 次 step_a_pure(0) ->', step_a_pure(0))
print('   第 3 次 step_a_pure(0) ->', step_a_pure(0))
print('   <- 同样的输入永远给同样的输出，和调用历史无关')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1 | `state = 0` | 全局状态 |
| 4-7 / 10-13 | `step_a` / `step_b` | 两个都要 `global state`，都改全局 |
| 16-20 | 连续调用三次 `step_a()` | **症状 1**：一模一样的调用，得到 1、2、3 —— 因为 `state` 在累积 |
| 23-27 | 换调用顺序 | **症状 2**：`step_b()` 前面接 `step_a()` 得 10，前面没人得 0 |
| 30-38 | `step_a_pure(s)` | **纯函数版本**：不读全局、不写全局，**只依赖参数、只通过返回值输出** |
| 40-42 | 连续调用三次 `step_a_pure(0)` | 永远是 1。**和调用历史无关** |

> ⚠️ **请注意「症状」这两个字的确切含义。**
> `step_a()` 本身并没有算错 —— 它的定义就是「全局 state 加一」。
> **问题是「它的输出取决于你看不见的历史」，而你从调用处完全看不出来。**
>
> 这才是课件说 `global` 会导致 "complex dependencies and harder-to-debug code" 的真正含义：
> **bug 不是「算错了」，而是「同样的代码在不同时刻给出不同答案」。**

> 💡 **「纯函数」是软件工程里最重要的概念之一**：
> **相同输入永远给相同输出，不碰外部状态。**
>
> 好处：
> 1. **可测试** —— 不用先摆好一堆全局变量
> 2. **可重排** —— 顺序无关
> 3. **可并行** —— 没有共享状态
> 4. **好读** —— 看函数签名就知道它要什么、给什么
>
> ⚠️ 写物理模拟时这条尤其重要：**你的「能量」「时间步」「随机数种子」如果全用全局变量，
> 想跑两组参数对比就得手动 reset，迟早出错。**

# §6 · 嵌套函数（P7–P9）

## 课件原文（P7）

- **Definition:** A nested function is a function **defined inside another function** (the parent function).
- The nested function **can only be called by the parent function**.
- **Why use nested functions?**
  - Useful when a task needs to be performed **multiple times within the parent function** but is **not required outside**.
  - Helps in organizing the code and preventing redundancy.

```python
def parent_function():
    def nested_function():
        # Code for nested function
    # Code that calls the nested function
```

## 课件原文（P8–P9）

**Memory and Scope in Nested Functions**

- A nested function has its **own separate memory block**, independent of the parent function.
- This allows variables with the same name to exist in both the parent and nested functions without conflict.
- **Practical Example:**

```python
def my_dist_xyz(x, y, z):
    def my_dist(x, y):
        return np.sqrt((x[0]-y[0])**2 + (x[1]-y[1])**2)
    d0 = my_dist(x, y)
    d1 = my_dist(x, z)
    d2 = my_dist(y, z)
    return [d0, d1, d2]
```

**Advantages of Using Nested Functions**

- **Code clarity:** Organizing complex functions into smaller, task-specific sub-functions reduces clutter and enhances readability.
- **Error reduction:** Nested functions reduce the risk of mistakes (like formula errors), as computations are **centralized within a single subfunction**.
- **Cleaner code:** Without a nested function, the parent function may become cluttered with **repeated logic**, increasing the chances of errors and making it harder to maintain.

## 人话版

**嵌套函数 = 「屋里的屋」。**

```python
def parent():          # 外层（父函数）
    def child():       # 内层（子函数），只有父函数能叫它
        ...
    child()            # 父函数里调用它
```

**为什么要把函数塞进函数里？** 课件给了答案，翻译成一句话：

> **「这个操作在父函数里要重复好几次，但外面的人根本用不着。」**

**课件 P9 那句话是全节的精华**：

> **Error reduction:** computations are **centralized within a single subfunction**.

**「把计算集中在一个子函数里」** —— 这句话值得展开。

假设你要算三个点两两之间的距离。**笨办法是把公式抄三遍**：

```python
d0 = np.sqrt((x[0]-y[0])**2 + (x[1]-y[1])**2)
d1 = np.sqrt((x[0]-z[0])**2 + (x[1]-z[1])**2)   # <- 抄一遍
d2 = np.sqrt((y[0]-z[0])**2 + (y[1]-z[1])**2)   # <- 再抄一遍
```

**问题在哪？** 如果公式要改（比如改成三维），**你得改三个地方，漏一个就是 bug**。

**用嵌套函数，公式只写一遍：**

```python
def my_dist_xyz(x, y, z):
    def my_dist(x, y):                     # <- 公式只在这里出现一次
        return np.sqrt((x[0]-y[0])**2 + (x[1]-y[1])**2)
    d0 = my_dist(x, y)                     # <- 三处调用，不可能抄错
    d1 = my_dist(x, z)
    d2 = my_dist(y, z)
    return [d0, d1, d2]
```

> 💡 **这叫「一次且仅一次」（DRY: Don't Repeat Yourself）原则。**
> **重复的代码是 bug 的温床** —— 改一处漏一处，是软件史上最常见的错误来源。

## §6.1 课件的 `my_dist_xyz`

````python
import numpy as np


def my_dist_xyz(x, y, z):
    def my_dist(x, y):
        return np.sqrt((x[0] - y[0])**2 + (x[1] - y[1])**2)

    d0 = my_dist(x, y)
    d1 = my_dist(x, z)
    d2 = my_dist(y, z)
    return [d0, d1, d2]


A = [0, 0]
B = [3, 0]
C = [0, 4]

print('三个点: A =', A, ' B =', B, ' C =', C)
print('两两距离 =', my_dist_xyz(A, B, C))
print()
print('  验证 AB：sqrt(3^2 + 0^2) =', np.sqrt(3**2 + 0**2))
print('  验证 AC：sqrt(0^2 + 4^2) =', np.sqrt(0**2 + 4**2))
print('  验证 BC：sqrt(3^2 + 4^2) =', np.sqrt(3**2 + 4**2), '  <- 3-4-5 直角三角形')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-5 | `def my_dist(x, y):` | **嵌套函数**：定义在 `my_dist_xyz` **里面**。注意它的函数体又缩进了一层 |
| 5 | `(x[0]-y[0])**2 + (x[1]-y[1])**2` | 两点距离公式。`x[0]`/`x[1]` 是第一个点的两个坐标 |
| 7-9 | `d0 = my_dist(x, y)` 等 | ⚠️ **公式只写了一遍，这里调用三次** |
| 10 | `return [d0, d1, d2]` | 注意这里返回的是**列表**（用了方括号），不是元组 |
| 13-15 | `A = [0,0]` 等 | 三个点用**列表**表示（所以才能用 `x[0]` 取坐标） |
| 18-21 | 验证 | A(0,0)、B(3,0)、C(0,4)：AB=3、AC=4、BC=5 |

> ⚠️ **一个细节：嵌套函数 `my_dist` 的参数也叫 `x, y`，和外层的 `x, y, z` 重名了！**
>
> 这**完全合法**，而且正是课件 P8 说的
> "This allows variables with the same name to exist in both the parent and nested functions without conflict"。
> 因为 `my_dist` 里的 `x, y` 是**它自己的局部变量**，和外层的是两回事。
>
> 💡 **但可读性上不推荐这么写。** 参数和外层重名，读代码的人得停下来想「这个 x 是哪个 x」。
> 更好的命名是 `def my_dist(p, q):`。**课件这里是为了教学才故意重名的。**

## §6.2 嵌套函数的作用域：子函数能看见父函数的变量

这是嵌套函数和「两个并列函数」最大的区别。

````python
print('--- 子函数能「看见」父函数的变量 ---')


def parent(v):
    scale = 100          # 父函数的局部变量

    def child(w):
        return w * scale   # 子函数直接用了父函数的 scale，没有当参数传！

    return child(v)


print('  parent(5) =', parent(5), '  <- 5 * 100')
print()
print('--- 但外面看不见 child ---')
try:
    child(1)
except NameError as e:
    print('  child(1) ->', type(e).__name__ + ':', e)
print()
print('--- 外面也看不见父函数里的 scale ---')
try:
    print(scale)
except NameError as e:
    print('  scale ->', type(e).__name__ + ':', e)
print()
print('--- 父函数的参数 v 和子函数的参数同名也互不干扰 ---')


def parent2(v):
    def child2(v):          # 同名参数！
        return v * 2
    print('  子函数里的 v =', child2(999))
    print('  回到父函数，v 还是', v)


parent2(7)
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-10 | `def parent(v):` | 父函数里定义了 `scale = 100`，子函数 `child` **直接用**了它，没有当参数传 |
| 7 | `return w * scale` | ⚠️ **`scale` 是从外层「捕获」进来的**。这叫**闭包（closure）** |
| 15-16 | 外面调 `child(1)` | ❌ `NameError`：子函数的名字**只活在父函数内部** |
| 23-24 | 外面调 `scale` | ❌ `NameError`：父函数的局部变量外面也看不见 |
| 28-33 | 同名参数 | 父的 `v` 和子的 `v` 是**两个不同的变量**，互不干扰 |

> 💡 **「子函数能读父函数的变量」叫闭包（closure）**，
> 这是 Python 里一个很强的能力（也是 lambda 和装饰器的基础）。
> **本课不深入，你只要知道「能看见」就够了。**
>
> ⚠️ **但「能读」不等于「能改」** —— 和 §5 的全局变量规则一样：
> 想在子函数里**改**父函数的变量，得用 `nonlocal` 关键字（这是 `global` 的兄弟）。
>
> 💡 **实用建议**：嵌套函数**最好只读外层的常量**（比如上面的 `scale`），
> 不要在外层和里层之间来回改状态。**要传的东西就用参数传。**

## §6.3 🧪 用嵌套函数消除重复：三维版本

课件说「公式集中在一个子函数里，改起来只改一处」。下面**真的改一次给你看** ——
把二维距离升级成三维，只需要动**一行**。

````python
def dist_pairs(points, n_dim):
    """算一堆点两两之间的距离（维度由参数决定）"""

    def dist(a, b):
        # 要改维度，只改这一行的 n_dim 来源即可；公式本身只写一次
        return float(np.sqrt(sum((a[i] - b[i])**2 for i in range(n_dim))))

    out = {}
    n = len(points)
    for i in range(n):
        for j in range(i + 1, n):
            out[(i, j)] = dist(points[i], points[j])
    return out


pts = [[0, 0, 0], [1, 0, 0], [0, 1, 0], [0, 0, 1]]
print('点集:', pts)
for k, v in dist_pairs(pts, 3).items():
    print(f'  点{k[0]} 到 点{k[1]} 的距离 = {v:.6f}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-6 | `def dist(a, b):` | 嵌套函数把「距离公式」**封装成一个东西** |
| 6 | `sum((a[i]-b[i])**2 for i in range(n_dim))` | 用生成器求和，**维度由外层参数 `n_dim` 决定** —— 这就是「只写一遍」的威力 |
| 11-13 | 双重循环 | `range(i + 1, n)` 保证每对点只算一次 |
| 10 | `out = {}` | 用**字典**存结果，键是 `(i, j)` 这对下标 —— 这正是 C2 §6 说的「元组可以当键」 |
| 17-19 | 输出 | 从原点出发的三个单位向量之间，距离都是 √2 ≈ 1.414214 |

> 💡 **对照一下：如果没有嵌套函数会怎样？**
> 你得在循环里直接写 `np.sqrt((a[0]-b[0])**2 + (a[1]-b[1])**2 + (a[2]-b[2])**2)`，
> 而且**每加一个维度都要改这行公式**，改错了还不会报错。
>
> ⚠️ **这就是课件说的 "computations are centralized within a single subfunction" 的实际含义**：
> **把「会变的东西」集中在一个地方，其他地方只负责调用。**

# §7 · lambda 匿名函数（P10–P12）

## 课件原文（P10–P11）

- **Definition:** A lambda function is an **anonymous function**, meaning it has no name and is typically used for short, simple operations.
- **Syntax:** Lambda functions are written using the `lambda` keyword followed by any number of arguments, and an expression to be evaluated.
- **Use Case:** These functions are commonly used when a simple function is needed for a single task, especially in scenarios such as sorting, filtering, or applying functions in a concise manner.

```python
lambda arguments: expression

square = lambda x: x**2
print(square(2))   # Output: 4
```

- **Lambda with Sorting:** sorting a list of tuples based on the second element:
  ```python
  sorted([(1, 2), (2, 0), (4, 1)], key=lambda x: x[1])
  # Output: [(2, 0), (4, 1), (1, 2)]
  ```
- **Lambda for Simple Math Operations:**
  ```python
  my_adder = lambda x, y: x + y
  print(my_adder(2, 4))   # Output: 6
  ```
- **Advantages:** Compact code. Useful in functions like `map()`, `filter()`, and `sorted()` where a small function is required.

## 课件原文（P12）

- **Limitations:**
  - Lambda functions are limited to **one expression** and are not suitable for complex operations.
  - Lack of a name makes **debugging harder** for more complicated functions.
- **Best Practices:**
  - Use lambda functions for **simple tasks** such as mathematical operations, data sorting, or filtering.
  - Avoid using them for large or complex functions where traditional function definitions are more appropriate.

## 人话版

**lambda = 「一次性的小函数」。**

把这两个写法**并排看**，立刻就懂了：

```python
# 普通写法：4 行
def square(x):
    return x ** 2

# lambda 写法：1 行
square = lambda x: x ** 2
```

**逐字对照：**

| 普通函数 | lambda |
|---|---|
| `def` | `lambda` |
| `square(x)`（名字 + 参数） | `x`（**只有参数，没有名字**） |
| `:` 换行 + `return x**2` | `:` 后面**直接写表达式** `x ** 2` |

> ⚠️ **lambda 体内不能写 `return`** ——
> 冒号后面那个表达式的**值**就是返回值。写 `lambda x: return x**2` 是语法错误。

**为什么叫「匿名」？** 因为 `lambda x: x**2` 本身**没有名字**。
上面那个 `square = ...` 是我们**事后给它贴的标签**，不是它自带的名字。

**那它到底有什么用？** 答案在**「一次性」**三个字：

> 有些函数**只在一个地方用一次**，为了它专门写一个 `def` 太啰嗦。
> 这时候 lambda 可以直接**塞进调用的括号里**，不用起名字。

课件给的两个例子正好是两种典型：

| 例子 | lambda 用在哪 |
|---|---|
| `sorted(..., key=lambda x: x[1])` | 作为**参数**塞进去（不用起名） |
| `my_adder = lambda x, y: x + y` | 赋给一个变量（其实**没必要**，见下面） |

> 💡 **关于第二个例子，说句实话**：
> `my_adder = lambda x, y: x + y` 这种写法**在 PEP 8（Python 官方风格指南）里是被明确不推荐的**。
> 官方原话是 "Always use a def statement instead of an assignment statement that binds a lambda expression to an identifier"。
>
> **理由**：给它起名字的那一刻，它就不「匿名」了，
> 那么用 `def my_adder(x, y): return x + y` **更清楚**（有名字、能写 docstring、报错时的 traceback 也更好看）。
>
> **lambda 真正的主场是「当参数传」**，也就是 P11 的第一个例子和 §8 的内容。

## §7.1 课件的三个例子

````python
print('=== 例 1：lambda 求平方 ===')
square = lambda x: x**2
print('  square(2) =', square(2))
print('  type(square) =', type(square).__name__, '  <- 它真的是个函数')
print('  square.__name__ =', repr(square.__name__), '  <- 名字是 <lambda>！')

print()
print('=== 例 2：lambda 排序 ===')
data = [(1, 2), (2, 0), (4, 1)]
print('  原始          :', data)
print('  按第 2 个元素排:', sorted(data, key=lambda x: x[1]))

print()
print('=== 例 3：lambda 做简单数学 ===')
my_adder = lambda x, y: x + y
print('  my_adder(2, 4) =', my_adder(2, 4))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 2 | `square = lambda x: x**2` | 定义一个「吃 x、吐 x²」的匿名函数，贴个标签叫 `square` |
| 3 | `square(2)` | → 4（和课件一致） |
| 4 | `type(square)` | ⚠️ **是 `function`，和 `def` 定义的完全同类**。lambda 不是「另一种东西」，只是另一种写法 |
| 5 | `square.__name__` | → `'<lambda>'`。**这就是「匿名」的字面证据** |
| 10 | `sorted(data, key=lambda x: x[1])` | **本节的重点**。`key` 参数要一个「把元素换算成排序依据」的函数，lambda 直接塞进去 |
| 15 | `my_adder = lambda x, y: x + y` | 两个参数 → 6（和课件一致）。但**这种写法不推荐**（见上） |

**例 2 逐项拆解**：`key=lambda x: x[1]` 的意思是「**排序时看每个元组的第 1 号元素**」。

| 元组 | `x[1]`（排序依据） |
|---|---|
| `(1, 2)` | 2 |
| `(2, 0)` | **0** ← 最小 |
| `(4, 1)` | 1 |

按 0、1、2 排 → `[(2, 0), (4, 1), (1, 2)]`（和课件一致）。

> ⚠️ **`key` 函数只负责「换算成什么来排」，不负责「比较」。**
> 换算完 Python 自己会排序。**所以 `lambda` 里只需要返回一个值就行。**
>
> 💡 **`sorted(data, key=...)` 是 lambda 出现频率最高的场合**，没有之一。
> 排序规则稍微复杂一点（按第 2 列、按字符串长度、按字典的某个键），就用它。

## §7.2 🧪 lambda 的四种真实用法

````python
import numpy as np

print('--- 用法 1：按「自定义规则」排序 ---')
words = ['banana', 'apple', 'kiwi', 'cherry']
print('  按字母          :', sorted(words))
print('  按长度          :', sorted(words, key=lambda w: len(w)))
print('  按最后一个字母  :', sorted(words, key=lambda w: w[-1]))

print()
print('--- 用法 2：排序字典列表（读数据时的常客）---')
students = [
    {'name': 'Alice', 'score': 88},
    {'name': 'Bob', 'score': 95},
    {'name': 'Carol', 'score': 72},
]
by_score = sorted(students, key=lambda s: s['score'], reverse=True)
for s in by_score:
    print(f"  {s['name']:<6} {s['score']}")

print()
print('--- 用法 3：map / filter（把函数作用到每个元素）---')
nums = [1, 2, 3, 4, 5, 6]
print('  list(map(lambda x: x**2, nums))       =', list(map(lambda x: x**2, nums)))
print('  list(filter(lambda x: x % 2 == 0, nums)) =', list(filter(lambda x: x % 2 == 0, nums)))
print('  （现在更推荐写推导式: [x**2 for x in nums] —— 见 C5）')

print()
print('--- 用法 4：多参数 / 默认参数也可以 ---')
g = lambda x, y=10: x * y
print('  g(5)    =', g(5), '   <- 用了默认的 y=10')
print('  g(5, 2) =', g(5, 2))
print()
print('--- 但不能有多条语句 ---')
try:
    # 注意：这里必须用 eval，因为语法错误发生在【编译】阶段，
    # 直接写出来会让整个 cell 都编译不过（连 try 都进不去）
    bad = eval("lambda x: (y = x + 1, y * 2)")
except SyntaxError as e:
    print('  想写多行 -> SyntaxError:', e)
print('  lambda 只能有【一个表达式】，要写多行请用 def')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 5-7 | `sorted(words, key=...)` | 同一个列表，**三种排序规则**，全靠换 key |
| 7 | `key=lambda w: len(w)` | 按长度排。**也可以直接写 `key=len`** —— 因为 `len` 本身就是个函数！ |
| 17 | `sorted(students, key=lambda s: s['score'], reverse=True)` | 从字典列表里按某个字段排序。**读 CSV/JSON 之后几乎必用** |
| 26 | `map(lambda x: x**2, nums)` | 对每个元素应用函数。⚠️ **`map` 返回的是迭代器，要用 `list()` 包一层才看得到内容** |
| 27 | `filter(lambda x: x % 2 == 0, nums)` | 留下使函数为真的元素（筛偶数） |
| 36-37 | `lambda x, y=10: x * y` | ⚠️ **lambda 也支持默认参数** |
| 41-47 | `eval("lambda x: (y = x + 1, ...)")` | ❌ 想在 lambda 里写赋值 -> `SyntaxError`。**这就是课件说的「只能有一个表达式」** |

> ⚠️ **为什么这里要用 `eval`？**
> 因为**语法错误发生在「编译」阶段，而不是「运行」阶段** ——
> 如果你把 `lambda x: (y = x + 1, ...)` 直接写在 cell 里，
> **整个 cell 根本编译不过**，连 `try` 都进不去（Jupyter 会直接拒绝运行这一格）。
> 用 `eval(字符串)` 才能把「编译」推迟到运行时，从而被 `except` 接住。
>
> 💡 **这条区别很重要**：**SyntaxError 是「话没说通」，TypeError 是「话说通了但做不了」。**
> 前者在运行前就拦住了，后者才可能在代码里被捕获。

**关于第 7 行那个小技巧**：`key=lambda w: len(w)` 和 `key=len` **完全等价** ——
因为你要的就是「用 `len` 这个函数去换算」。**凡是能直接用一个现成函数的，就别包 lambda。**

> ⚠️ **`map` 和 `filter` 现在不推荐了。**
> Python 官方风格上，**列表推导式更清楚**（C5 会讲）：
>
> | 老写法 | 新写法 |
> |---|---|
> | `list(map(lambda x: x**2, nums))` | `[x**2 for x in nums]` |
> | `list(filter(lambda x: x%2==0, nums))` | `[x for x in nums if x%2==0]` |
>
> **但 `sorted(key=lambda ...)` 依然是 lambda 的主场**，没有替代品。

## §7.3 课件的 Limitations：什么时候**不要**用 lambda

课件的两条限制，翻译 + 补充：

| 课件原文 | 人话 | 后果 |
|---|---|---|
| limited to **one expression** | 只能写一个表达式 | 想写 `if`/`for`/`=` 就不行（除非用三元表达式，见 C4） |
| **Lack of a name** makes debugging harder | 报错里显示 `<lambda>`，不知道是哪个 | traceback 里全是 `<lambda>`，一排下来根本分不清 |

**第二条到底有多痛？看下面这个实测。**

````python
import traceback


def named_func(x):
    return x / 0        # 故意除零


anon = lambda x: x / 0  # 同样的事，但没名字


print('--- 有名字的函数，报错长这样 ---')
try:
    named_func(5)
except ZeroDivisionError:
    tb = traceback.format_exc().strip().splitlines()
    for line in tb[-4:]:
        print('  ' + line)

print()
print('--- lambda 的报错，长这样 ---')
try:
    anon(5)
except ZeroDivisionError:
    tb = traceback.format_exc().strip().splitlines()
    for line in tb[-4:]:
        print('  ' + line)

print()
print('  差别就在那一行: named_func  vs  <lambda>')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4-5 | `def named_func(x):` | 一个**有名字**的函数，函数体里故意除零 |
| 8 | `anon = lambda x: x / 0` | 同样的错误，但它是 lambda |
| 13-17 | 打印 traceback 的**最后几行** | `traceback.format_exc()` 拿到完整报错文本，`splitlines()` 切成行，取最后 4 行 |
| 22-25 | 同上 | |

**关键差别**：报错里那一行会写 `in named_func` vs `in <lambda>`。

**一个 `<lambda>` 还好，十个呢？** 如果你的代码里有五个 lambda 都报 `in <lambda>`，
你根本不知道是哪一个出的问题。

> ⚠️ **实用判据：这个函数会不会出现在报错里？**
>
> | 情况 | 建议 |
> |---|---|
> | 一行能写完、只在一处用、**逻辑显然** | lambda ✓ |
> | 有任何计算步骤、会被复用到别处 | `def` ✓ |
> | 可能会报错、需要调试 | `def` ✓（除非你非常确定） |
>
> 💡 一句话：**lambda 是给「一眼就懂」的场合用的。看不懂就写 `def`。**

# §8 · 函数作为参数（P13–P15）

## 课件原文（P13）

- In Python, functions are treated as **first-class objects**, meaning they can be **passed as arguments to other functions**.
- This capability allows for greater flexibility in coding, making functions **interchangeable**.
- **Example:** The `max` function can be assigned to a variable `f`, and using `f` is equivalent to calling the `max` function.
  ```python
  f = max
  print(f([2, 3, 5]))   # Output: 5
  ```

## 课件原文（P14–P15）

- You can define custom functions that **accept other functions as arguments**.
- **Example:** A function `my_fun_plus_one` takes a function `f` and a value `x` as input, applies `f` to `x`, and adds 1 to the result.
  ```python
  import numpy as np

  def my_fun_plus_one(f, x):
      return f(x) + 1

  print(my_fun_plus_one(np.sin, np.pi/2))   # Output: 2.0
  print(my_fun_plus_one(np.sqrt, 25))       # Output: 6.0
  ```
- **Lambda functions** can also be passed as arguments:
  ```python
  print(my_fun_plus_one(lambda x: x + 2, 2))   # Output: 5
  ```

## 人话版

**「first-class objects（一等对象）」是什么意思？**

一个东西是「一等对象」，意思是它**和数字、字符串享受同等待遇**：

| 能做吗 | 数字 `5` | 函数 `max` |
|---|---|---|
| 赋给变量 | `a = 5` | `f = max` |
| 放进列表 | `[5, 6]` | `[max, min]` |
| 当参数传 | `len([5])` | `sorted(x, key=len)` |
| 当返回值 | `return 5` | `return max` |

**函数能做到这四件事，就叫「一等对象」。** Python 里函数全都能做到。

**这就是「把函数当参数传」的全部秘密** —— 没什么玄机，函数就是个普通的值。

**课件的 `my_fun_plus_one` 是本节的样板**：

```python
def my_fun_plus_one(f, x):
    return f(x) + 1
#          ^^^^
#          参数 f 是个【函数】，后面直接调用它
```

**看第 2 行那个 `f(x)`**：`f` 是我们**传进来的**东西，
我们不知道它是 sin 还是 sqrt 还是别的，**但我们可以调用它。**

| 调用 | `f` 是什么 | 结果 |
|---|---|---|
| `my_fun_plus_one(np.sin, np.pi/2)` | `np.sin` | sin(π/2) + 1 = 1 + 1 = **2.0** |
| `my_fun_plus_one(np.sqrt, 25)` | `np.sqrt` | √25 + 1 = 5 + 1 = **6.0** |
| `my_fun_plus_one(lambda x: x + 2, 2)` | 一个 lambda | (2+2) + 1 = **5** |

> 💡 **为什么这个能力很重要？**
> 因为它把「**做什么**」和「**对什么做**」分开了。
>
> `my_fun_plus_one` 负责「加 1」，至于先做什么运算，由**调用者**决定。
> **这就是「可复用」的最高形态：一个函数能适配无数种具体需求。**
>
> 📎 你其实早就在用了：`sorted(data, key=len)` 里的 `key`、
> C1 里 `math.isclose` 的用法，都是这个机制。

## §8.1 课件三个例子

````python
import numpy as np

print('=== 例 1：把函数赋给变量（P13）===')
f = max
print('  f = max;  f([2, 3, 5]) =', f([2, 3, 5]))
print('  f is max ->', f is max, '  <- 就是同一个对象，不是复制')

print()
print('=== 例 2：自定义函数接受函数参数（P14）===')


def my_fun_plus_one(f, x):
    return f(x) + 1


print('  my_fun_plus_one(np.sin, np.pi/2) =', my_fun_plus_one(np.sin, np.pi / 2))
print('  my_fun_plus_one(np.sqrt, 25)     =', my_fun_plus_one(np.sqrt, 25))

print()
print('=== 例 3：lambda 当参数（P15）===')
print('  my_fun_plus_one(lambda x: x + 2, 2) =', my_fun_plus_one(lambda x: x + 2, 2))
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4 | `f = max` | ⚠️ **注意没有括号！** `max` 是函数本身，`max([2,3,5])` 才是调用 |
| 5 | `f([2, 3, 5])` | 用 `f` 调用 → 5（和课件一致） |
| 6 | `f is max` | → `True`。**`f` 不是 `max` 的副本，而是同一个对象** |
| 12-13 | `def my_fun_plus_one(f, x):` | 参数 `f` 是个**函数**。函数体里 `f(x)` 就是「调用传进来的那个函数」 |
| 16-17 | 传 `np.sin` / `np.sqrt` | ⚠️ **传的是函数名，不带括号**。写了括号就变成「把 sin(π/2) 的结果传进去」了 |
| 21 | `lambda x: x + 2` | 直接把一个匿名函数塞进去，不用起名字 |

**三个输出全部和课件一致**：2.0、6.0、5。

> ⚠️ **本节最大的坑：传函数时忘/多写括号。**
>
> | 写法 | 传的是什么 | 结果 |
> |---|---|---|
> | `my_fun_plus_one(np.sin, 1)` | **函数本身** ✓ | sin(1) + 1 |
> | `my_fun_plus_one(np.sin(1), 1)` | sin(1) 的**结果**（一个数字） ✗ | `TypeError: 'numpy.float64' object is not callable` |
>
> **报错里的 `not callable`（不可调用）就是「你给的不是函数」的意思。**
> 以后看到这个词，第一反应就是「我是不是多写了括号」。

## §8.2 🧪 这个机制能干什么：一个通用「数值求导器」

把函数当参数传，最实用的场合是**写通用工具**。下面这个例子
是**物理学里最常用的数值方法**之一：中心差分求导。

````python
def derivative(f, x, h=1e-5):
    """用中心差分法求 f 在 x 处的导数：f'(x) ≈ [f(x+h) - f(x-h)] / (2h)"""
    return (f(x + h) - f(x - h)) / (2 * h)


print('--- 拿几个已知导数的函数验证 ---')
print(f'  d/dx sin(x) 在 x=0    = {derivative(np.sin, 0):.10f}   （理论值 cos(0) = 1）')
print(f'  d/dx sin(x) 在 x=pi/2 = {derivative(np.sin, np.pi/2):.10f}   （理论值 cos(pi/2) = 0）')
print(f'  d/dx x^2    在 x=3    = {derivative(lambda x: x**2, 3):.10f}   （理论值 2x = 6）')
print(f'  d/dx exp(x) 在 x=1    = {derivative(np.exp, 1):.10f}   （理论值 e = {np.e:.10f}）')
print()
print('--- 同一个 derivative，能求任何函数的导数 ---')
funcs = [('sin', np.sin, 0.0, 1.0),
         ('cos', np.cos, 0.0, 0.0),
         ('exp', np.exp, 0.0, 1.0),
         ('log', np.log, 1.0, 1.0)]
for name, fn, x0, expect in funcs:
    got = derivative(fn, x0)
    print(f'  {name:<4} 在 x={x0} 的导数 = {got: .8f}   理论值 = {expect: .8f}')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 1-3 | `def derivative(f, x, h=1e-5):` | **`f` 是函数参数**；`h` 是步长，用了**默认参数**（§3 学的） |
| 3 | `(f(x + h) - f(x - h)) / (2 * h)` | **中心差分公式**。这个函数**完全不知道 `f` 是什么**，但它能算任何函数的导数 |
| 7-10 | 验证四个 | 和理论值对照，误差在 1e-10 量级 |
| 13-18 | 循环调用 | ⚠️ **把函数放进列表里**（元组 `('sin', np.sin, 0.0, 1.0)`）—— 这再次证明「函数是普通的值」 |

> 💡 **这就是「函数作为参数」的威力：写一次，适配所有函数。**
> 如果不用这个机制，你得为每个函数写一个 `derivative_of_sin`、`derivative_of_exp`……
>
> ⚠️ **一个物理人必须知道的细节**：`h=1e-5` 这个步长是**权衡的结果**。
>
> | `h` 太大 | `h` 太小 |
> |---|---|
> | 截断误差大（公式本身是近似） | 舍入误差大（两个相近的数相减） |
>
> 想深入理解这件事，看 **C9（数字表示与舍入误差）**。
> 那里的结论是：**中心差分的最佳步长大约是 1e-5 量级，不是越小越好。**
>
> 📎 试试把 `h` 改成 `1e-1` 和 `1e-15`，看误差怎么变。**这是 C9 的预习题。**

## §8.3 🧪 三个「高阶函数」：把函数当参数的内置工具

Python 自带了几个「接收函数」的工具。课件没列，但它们是这个机制最常见的出口。

````python
nums = [1, -2, 3, -4, 5]

print('--- sorted(key=...)  ：按自定义规则排序 ---')
print('  按绝对值排序:', sorted(nums, key=abs))
print('  按「离 3 的距离」排序:', sorted(nums, key=lambda x: abs(x - 3)))

print()
print('--- max / min(key=...) ：按自定义规则找极值 ---')
print('  绝对值最大的数:', max(nums, key=abs))
print('  直接 max     :', max(nums), '  <- 不加 key 就是按数值大小')

print()
print('--- any / all ：把一组判断合起来 ---')
print('  any(x > 4 for x in nums) =', any(x > 4 for x in nums), '  <- 至少一个满足')
print('  all(x > 0 for x in nums) =', all(x > 0 for x in nums), '  <- 全部满足')

print()
print('--- 用 key=abs 做一次「按绝对值排序」，并保持稳定 ---')
pairs = [(-3, 'a'), (2, 'b'), (-1, 'c'), (4, 'd')]
print('  原始            :', pairs)
print('  按绝对值排序     :', sorted(pairs, key=lambda p: abs(p[0])))
print('  （第 2 个元素完全没参与排序，只是被带着走）')
````

**📖 这一格在做什么**

| 行 | 代码 | 人话 |
|---|---|---|
| 4 | `sorted(nums, key=abs)` | ⚠️ **`abs` 是内置函数，直接传就行**，不用包 lambda。按绝对值从小到大 |
| 5 | `key=lambda x: abs(x - 3)` | 排序依据是「离 3 有多远」。这个规则**没有现成函数**，所以必须用 lambda |
| 9 | `max(nums, key=abs)` | 找「绝对值最大」的那个 —— 返回的是**元素本身**（5），不是它的绝对值 |
| 10 | `max(nums)` | 不加 `key` 就是普通的比较 → 5 |
| 14-15 | `any(...)` / `all(...)` | ⚠️ **这两个收的是「一堆 True/False」**，所以里面通常写生成器表达式（C5 会讲） |
| 19-21 | 排序元组列表 | 排序依据只有第一个元素，第二个元素被带着走 |

> 💡 **`key` 系列是「函数作为参数」最高频的应用**，四个要记住：
>
> | 函数 | 用途 |
> |---|---|
> | `sorted(data, key=f)` | 按 `f` 排 |
> | `max(data, key=f)` | 按 `f` 找最大 |
> | `min(data, key=f)` | 按 `f` 找最小 |
> | `list.sort(key=f)` | 就地排（C2 §3.6） |
>
> ⚠️ **`any` / `all` 的经典陷阱**：`all([])` 返回 **`True`**（空集「全部满足」是真空真），
> `any([])` 返回 **`False`**。写校验逻辑时别忘了这个边界情况。

# §9 · 一张表收尾

## 这一章你到底学会了什么

| 主题 | 最小必背 |
|---|---|
| 定义 | `def 名字(参数):` + **缩进的函数体** + 可选 `return` |
| 调用 | **有括号才是调用**；`print(f)` 打出的是 `<function ...>` |
| 多返回值 | 其实是**返回元组**，靠解包接住 |
| 默认参数 | `def f(x, y=10)`；**默认值只算一次**，别用可变对象 |
| 局部变量 | 函数里的名字**出了函数就消失**；参数也是局部的 |
| 全局变量 | **只读**随便读；**要重新赋值**必须 `global` |
| 嵌套函数 | 屋里的屋；能读父函数的变量（闭包），外面看不见 |
| lambda | `lambda x: 表达式`；**只能一个表达式**；主战场是 `sorted(key=...)` |
| 函数作为参数 | 函数是**一等对象**，能赋值、能传参、能放列表 |

## 四个坑（本章全部实测过）

1. **`return` 忘记写** → 函数默默返回 `None`，后面所有计算都变成「`None` 的运算」。
2. **默认参数用了 `[]`** → 所有调用**共享同一个列表**，数据会莫名其妙累积。用 `None` 代替。
3. **在函数里改全局变量却忘了 `global`** → `UnboundLocalError`，而且报错点在赋值之前。
4. **传函数时多写了括号** → `TypeError: 'xxx' object is not callable`。

## 四个好习惯

1. **每个函数写一行 docstring** —— 它会被 `?` 显示出来，不白写。
2. **函数只通过参数收、通过 `return` 给** —— 别偷偷读写全局变量（§5.4 实测过顺序依赖的可怕）。
3. **重复的公式一定抽成函数** —— 课件 P9 说的 "centralized within a single subfunction" 就是这个。
4. **不知道用 lambda 还是 `def` 时，用 `def`** —— 更清楚、能写文档、报错更好看。
